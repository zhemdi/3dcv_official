# Повторение: Линейная модель камеры

$\Large \mathbf{x_w}$ – координаты точки сцены $P$ в мировой системе координат  
$\Large \mathbf{x_с}$ – координаты $P$ в системе координат, связанной с камерой  
$\Large (x_i, y_i)$ – координаты проекции $P$ на плоскость изображения в системе координат связяанной с этой плоскостью  

$$\Large \mathbf{x_с} = {R} (\mathbf{x_w} - \mathbf{c_w}) = {R} \mathbf{x_w} + \mathbf{t}, \qquad \Large \mathbf{t} = \mathbf{-R}\mathbf{c_w}$$


$$\Large
R=
\begin{bmatrix}
r_{11} & r_{12} & r_{13}\\
r_{21} & r_{22} & r_{23}\\
r_{31} & r_{32} & r_{33}
\end{bmatrix}
{\normalsize \text{– матрица поворота}}$$

$$\Large \text{Перспективная проекция:}$$


$$\Large \frac{\Large x_i}{\Large f} = \frac{\Large x_c}{\Large z_c}, \qquad \Large \frac{\Large y_i}{\Large f} = \frac{\Large y_c}{\Large z_c}$$

$$\Large {x_i} = {\Large f}\frac{\Large x_c}{\Large z_c}, \qquad \Large {y_i} = {\Large f}\frac{\Large y_c}{\Large z_c}$$

$\Large {(u, v)} \text{ – координаты } (x_i, y_i) \text{ в пикселях}$


$\Large {(o_x, o_y)} \text{ – Prinipal point – точка, в которой оптическая ось линзы пересекает сенсор}$

$\Large {m_x} \text{ и } {m_y} \text{ –  плотности пикселей по } \hat{x}_i \text{ и } \hat{y}_i ( pixels / mm)$



$$\Large {u} = {m_x}{x_i} + {o_x} = {m_x}{f} \Large \frac{\Large x_c}{\Large z_c} + {o_x} = {f_x} \Large \frac{\Large x_c}{\Large z_c} + {o_x},$$

$$\Large {v} = {m_y}{y_i} + {o_y} = {m_y}{f} \Large \frac{\Large y_c}{\Large z_c} + {o_y} = {f_y} \Large \frac{\Large y_c}{\Large z_c} + {o_y}.$$

$\Large ({f_x}, {f_y}, {o_x}, {o_y}) \text{ – Intrinsic parameters of the camera.}$

$\Large {R}, \mathbf{t} \text{ – Extrinsic parameters of the camera.}$


$$\Large \text{Однородные координаты}$$

$\Large \text{Проекция } (x_c, y_c, z_c) \to (u, v) \text{ нелинейна (деление на } z_c\text{). Чтобы записать её матрично, перейдём к однородным координатам:}$

$$\Large (u, v) \;\to\; \tilde{\mathbf{u}} = \begin{bmatrix} \tilde{u} \\ \tilde{v} \\ \tilde{w} \end{bmatrix} = \begin{bmatrix} z_c u \\ z_c v \\ z_c \end{bmatrix} \equiv \begin{bmatrix} u \\ v \\ 1 \end{bmatrix}, \qquad u = \frac{\tilde{u}}{\tilde{w}}, \quad v = \frac{\tilde{v}}{\tilde{w}}$$

$\Large \text{Однородные векторы, отличающиеся ненулевым множителем, задают одну и ту же точку.}$

$$\Large \mathbf{x_w} = \begin{bmatrix} x_w \\ y_w \\ z_w \end{bmatrix} \;\to\; \tilde{\mathbf{x}}_\mathbf{w} = \begin{bmatrix} x_w \\ y_w \\ z_w \\ 1 \end{bmatrix}, \qquad \mathbf{x_c} \;\to\; \tilde{\mathbf{x}}_\mathbf{c} = \begin{bmatrix} x_c \\ y_c \\ z_c \\ 1 \end{bmatrix}$$

$$\Large \text{Матрица внутренних параметров (Intrinsic matrix)}$$

$$\Large
\begin{bmatrix} \tilde{u} \\ \tilde{v} \\ \tilde{w} \end{bmatrix}
=
\begin{bmatrix} z_c u \\ z_c v \\ z_c \end{bmatrix}
=
\begin{bmatrix} f_x x_c + o_x z_c \\ f_y y_c + o_y z_c \\ z_c \end{bmatrix}
=
\begin{bmatrix}
f_x & 0 & o_x & 0\\
0 & f_y & o_y & 0\\
0 & 0 & 1 & 0
\end{bmatrix}
\begin{bmatrix} x_c \\ y_c \\ z_c \\ 1 \end{bmatrix}
= \big[\, K \mid \mathbf{0} \,\big]\, \tilde{\mathbf{x}}_\mathbf{c}$$

$$\Large
K =
\begin{bmatrix}
f_x & 0 & o_x\\
0 & f_y & o_y\\
0 & 0 & 1
\end{bmatrix}
{\normalsize \text{– матрица внутренних параметров (calibration matrix)}}$$

$$\Large \text{Матрица внешних параметров (Extrinsic matrix)}$$

$$\Large \mathbf{x_c} = R\,\mathbf{x_w} + \mathbf{t}
\quad \Longrightarrow \quad
\begin{bmatrix} x_c \\ y_c \\ z_c \\ 1 \end{bmatrix}
=
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & t_x\\
r_{21} & r_{22} & r_{23} & t_y\\
r_{31} & r_{32} & r_{33} & t_z\\
0 & 0 & 0 & 1
\end{bmatrix}
\begin{bmatrix} x_w \\ y_w \\ z_w \\ 1 \end{bmatrix},
\qquad
\tilde{\mathbf{x}}_\mathbf{c} =
\begin{bmatrix} R & \mathbf{t} \\ \mathbf{0}^T & 1 \end{bmatrix}
\tilde{\mathbf{x}}_\mathbf{w}$$

$$\Large \text{Итоговая формула проекции}$$

$$\Large
\tilde{\mathbf{u}} = \big[\, K \mid \mathbf{0} \,\big]
\begin{bmatrix} R & \mathbf{t} \\ \mathbf{0}^T & 1 \end{bmatrix}
\tilde{\mathbf{x}}_\mathbf{w}
= K \big[\, R \mid \mathbf{t} \,\big]\, \tilde{\mathbf{x}}_\mathbf{w}$$

$$\Large
z_c
\begin{bmatrix} u \\ v \\ 1 \end{bmatrix}
=
\underbrace{
\begin{bmatrix}
f_x & 0 & o_x\\
0 & f_y & o_y\\
0 & 0 & 1
\end{bmatrix}}_{K\ \text{(intrinsic)}}
\underbrace{
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & t_x\\
r_{21} & r_{22} & r_{23} & t_y\\
r_{31} & r_{32} & r_{33} & t_z
\end{bmatrix}}_{[R \mid \mathbf{t}]\ \text{(extrinsic)}}
\begin{bmatrix} x_w \\ y_w \\ z_w \\ 1 \end{bmatrix}$$

$$\Large \tilde{\mathbf{u}} = P\,\tilde{\mathbf{x}}_\mathbf{w}, \qquad P = K \big[\, R \mid \mathbf{t} \,\big] {\normalsize \text{ – матрица проекции камеры } (3 \times 4)}$$

$\Large \textbf{Вопрос: } \text{можно ли, имея откалиброванную камеру, по изображению сцены найти 3D координаты точек сцены?}$

$$\Large \text{Простая стереопара}$$

$\Large \text{Две одинаковые камеры (одна и та же матрица } K\text{) с параллельными оптическими осями.}$  
$\Large \text{Мировая система координат совпадает с системой координат левой камеры, правая камера сдвинута вдоль оси } x \text{ на } b\text{:}$

$$\Large \mathbf{c}_l = (0, 0, 0), \qquad \mathbf{c}_r = (b, 0, 0), \qquad b \text{ – база (baseline)}$$

$\Large \text{Точка сцены } P = (x, y, z) \text{ проецируется в } (u_l, v_l) \text{ на левом изображении и в } (u_r, v_r) \text{ на правом.}$

$$\Large \text{Левая камера: } \quad u_l = f_x \frac{x}{z} + o_x, \qquad v_l = f_y \frac{y}{z} + o_y$$

$$\Large \text{Правая камера: } \quad u_r = f_x \frac{x - b}{z} + o_x, \qquad v_r = f_y \frac{y}{z} + o_y$$

$\Large \text{Отсюда } v_l = v_r \text{ – соответствующие точки лежат на одной строке изображения.}$

$$\Large \textbf{Диспаратность (disparity): } \quad d = u_l - u_r = f_x \frac{x}{z} - f_x \frac{x - b}{z} = \frac{f_x\, b}{z}$$

$\Large \text{Диспаратность обратно пропорциональна глубине: чем дальше точка, тем меньше } d\text{.}$

$$\Large \text{Триангуляция:}$$

$$\Large z = \frac{f_x\, b}{u_l - u_r}, \qquad
x = \frac{z\,(u_l - o_x)}{f_x} = \frac{b\,(u_l - o_x)}{u_l - u_r}, \qquad
y = \frac{z\,(v_l - o_y)}{f_y} = \frac{b\, f_x\,(v_l - o_y)}{f_y\,(u_l - u_r)}$$

# Epipolar geometry

# 1. Epipoles, Epipolar plane, Epipolar Line

$\Large \mathbf{t}, R$ – вектор сдвига и матрица поворота правой камеры относительно левой  
$\Large \mathbf{x_l}$ – координаты точки $P$ сцены в системе координат левой камеры  
$\Large \mathbf{x_r}$ – координаты точки $P$ сцены в системе координат правой камеры  
$\Large \mathbf{e_l}, \mathbf{e_r}$ – **эпиполи** – точки на плоскостях изображений, в которых виден центр другой камеры.
$\mathbf{e_l}$ и $\mathbf{e_r}$ уникальны для данной стереопары.

**Эпиполярная плоскость** для точки $P$ – плоскость, в которой лежат центры камер, эпиполи и точка сцены $P$.
Для каждой точки сцены своя эпиполярная плоскость.

**Эпиполярные линии** – линии пересечения эпиполярной плоскости с плоскостями изображений.

**Суть эпиполярных линий:** для каждой точки на одном изображении соответствующая точка на втором изображении
лежит на эпиполярной линии. Это свойство полезно тем, что сводит поиск соответствующих точек к **одномерному** поиску.

# 2. Epipolar constraint and Essential Matrix

**Цель:** найти условие, связывающее координаты одной и той же точки $P$ в двух камерах, которое зависит только от взаимного положения камер.

Переход из системы координат правой камеры в систему координат левой:
$$\Large \mathbf{x_l} = R\,\mathbf{x_r} + \mathbf{t}$$
$\mathbf{t}$ – центр правой камеры в системе координат левой, т.е. вектор базы $O_l O_r$.

### Шаг 1. Нормаль к эпиполярной плоскости

Векторы $\mathbf{t} = \overrightarrow{O_l O_r}$ и $\mathbf{x_l} = \overrightarrow{O_l P}$ оба лежат в эпиполярной плоскости,
значит их векторное произведение – нормаль к ней:
$$\Large \mathbf{n} = \mathbf{t} \times \mathbf{x_l}$$

### Шаг 2. Условие компланарности

$\mathbf{x_l}$ лежит в эпиполярной плоскости, поэтому он перпендикулярен нормали:
$$\Large \mathbf{x_l} \cdot (\mathbf{t} \times \mathbf{x_l}) = 0$$

### Шаг 3. Векторное произведение как умножение на матрицу

$$\Large \mathbf{t} \times \mathbf{a} = [\mathbf{t}]_\times\, \mathbf{a}, \qquad
[\mathbf{t}]_\times =
\begin{bmatrix}
0 & -t_z & t_y\\
t_z & 0 & -t_x\\
-t_y & t_x & 0
\end{bmatrix}
{\normalsize \text{– кососимметричная матрица}}$$

Тогда условие из шага 2:
$$\Large \mathbf{x_l}^T\, [\mathbf{t}]_\times\, \mathbf{x_l} = 0$$

### Шаг 4. Подставляем $\mathbf{x_l} = R\,\mathbf{x_r} + \mathbf{t}$ в правый множитель

$$\Large \mathbf{x_l}^T\, [\mathbf{t}]_\times \left(R\,\mathbf{x_r} + \mathbf{t}\right)
= \mathbf{x_l}^T\, [\mathbf{t}]_\times R\, \mathbf{x_r} + \mathbf{x_l}^T \underbrace{(\mathbf{t} \times \mathbf{t})}_{=\,\mathbf{0}}
= 0$$

$$\Large \mathbf{x_l}^T\, [\mathbf{t}]_\times R\, \mathbf{x_r} = 0$$

### Результат: эпиполярное ограничение

$$\Large \boxed{\ \mathbf{x_l}^T E\, \mathbf{x_r} = 0, \qquad E = [\mathbf{t}]_\times R\ }$$

$E$ – **существенная матрица** (Essential Matrix).

**Свойства:**
* $E$ – матрица $3\times3$, зависит **только от внешних параметров** $R, \mathbf{t}$;
* $\operatorname{rank} E = 2$, так как $\operatorname{rank}\,[\mathbf{t}]_\times = 2$;
* 5 степеней свободы: 3 – поворот $R$, 2 – направление $\mathbf{t}$. Уравнение однородное, поэтому масштаб $\mathbf{t}$ (и $E$) по двум видам определить нельзя;
* по той же причине ограничение выполняется для **любых** точек на лучах $O_lP$ и $O_rP$, например для $\mathbf{x}/z$.
  Но $\mathbf{x_l}, \mathbf{x_r}$ – координаты в системе камеры (в метрах), а измеряем мы пиксели $(u, v)$ →
  следующий шаг: связать пиксели с $E$ через внутренние параметры $K$.

# 3. Fundamental Matrix

**Проблема:** в ограничении $\mathbf{x_l}^T E\, \mathbf{x_r} = 0$ стоят координаты точки в системе камеры, а на изображении мы измеряем
только пиксели $(u, v)$. Выразим $\mathbf{x_l}$ и $\mathbf{x_r}$ через пиксельные координаты и внутренние параметры камер.

Перспективная проекция:

$\Large {u} = {f_x} \Large \frac{\Large x_c}{\Large z_c} + {o_x},$

$\Large {v} = {f_y} \Large \frac{\Large y_c}{\Large z_c} + {o_y}.$

### Шаг 1. Проекция в матричном виде

Умножим оба уравнения на $z_c$ и добавим тождество $z_c = z_c$:
$$\Large
\begin{aligned}
z_c\, u &= f_x\, x_c + o_x\, z_c\\
z_c\, v &= f_y\, y_c + o_y\, z_c\\
z_c &= z_c
\end{aligned}
\qquad\Longrightarrow\qquad
z_c
\begin{bmatrix} u\\ v\\ 1 \end{bmatrix}
=
\underbrace{\begin{bmatrix}
f_x & 0 & o_x\\
0 & f_y & o_y\\
0 & 0 & 1
\end{bmatrix}}_{K}
\begin{bmatrix} x_c\\ y_c\\ z_c \end{bmatrix}$$

$K$ – матрица внутренних параметров (3×3, обратима).

### Шаг 2. Выражаем $\mathbf{x_l}$ и $\mathbf{x_r}$ через пиксели

Для левой и правой камер (у каждой своя $K$):
$$\Large
\mathbf{x_l} = z_l\, K_l^{-1} \begin{bmatrix} u_l\\ v_l\\ 1 \end{bmatrix},
\qquad
\mathbf{x_r} = z_r\, K_r^{-1} \begin{bmatrix} u_r\\ v_r\\ 1 \end{bmatrix}$$

### Шаг 3. Подставляем в эпиполярное ограничение

Используем $\mathbf{x_l}^T = z_l\, [u_l\ \ v_l\ \ 1]\, K_l^{-T}$:
$$\Large
z_l\, [u_l\ \ v_l\ \ 1]\; K_l^{-T} E\, K_r^{-1}
\begin{bmatrix} u_r\\ v_r\\ 1 \end{bmatrix} z_r = 0$$

Точка $P$ находится перед камерами: $z_l \ne 0,\ z_r \ne 0$ → делим на $z_l z_r$:
$$\Large
[u_l\ \ v_l\ \ 1]\; K_l^{-T} E\, K_r^{-1}
\begin{bmatrix} u_r\\ v_r\\ 1 \end{bmatrix} = 0$$

### Результат: фундаментальная матрица

$$\Large \boxed{\
[u_l\ \ v_l\ \ 1]\; F
\begin{bmatrix} u_r\\ v_r\\ 1 \end{bmatrix} = 0,
\qquad F = K_l^{-T} E\, K_r^{-1} = K_l^{-T}\, [\mathbf{t}]_\times R\; K_r^{-1}\ }$$

$F$ – **фундаментальная матрица** (Fundamental Matrix). Обратно: $\ E = K_l^T F K_r$.

**Свойства:**
* $F$ связывает **пиксельные** координаты и зависит и от внутренних ($K_l, K_r$), и от внешних ($R, \mathbf{t}$) параметров;
* $\operatorname{rank} F = 2$, так как $\operatorname{rank} E = 2$, а $K_l, K_r$ обратимы;
* 7 степеней свободы: 9 элементов − 1 (масштаб, уравнение однородное) − 1 ($\det F = 0$);
* **эпиполярные линии:** для точки $\tilde{\mathbf{u}}_l = (u_l, v_l, 1)^T$ соответствующая точка правого изображения лежит на прямой
  $\ \mathbf{l_r} = F^T \tilde{\mathbf{u}}_l\ $ (т.е. $a u_r + b v_r + c = 0$), и наоборот: $\ \mathbf{l_l} = F\, \tilde{\mathbf{u}}_r$;
* **эпиполи:** все эпиполярные линии проходят через эпиполь, поэтому $\ F\, \mathbf{e_r} = 0,\ \ F^T \mathbf{e_l} = 0$;
* главное: $F$ можно оценить **только по соответствиям пикселей**, не зная ни $K$, ни $R, \mathbf{t}$ →
  дальше найдём $F$ по парам точек, а затем $E = K_l^T F K_r$ и из неё $R, \mathbf{t}$.

## 1. Эпиполярное ограничение как линейная система $A\mathbf{f} = 0$

Пиксельные координаты точек $\mathbf{u}_l = (u_l, v_l)^T$, $\mathbf{u}_r = (u_r, v_r)^T$ переводим в **однородные**:
$$\tilde{\mathbf{u}}_l = \begin{bmatrix} u_l \\ v_l \\ 1 \end{bmatrix}, \qquad
\tilde{\mathbf{u}}_r = \begin{bmatrix} u_r \\ v_r \\ 1 \end{bmatrix}.$$

Раскроем $\tilde{\mathbf{u}}_l^T F\, \tilde{\mathbf{u}}_r = 0$ поэлементно:
$$\sum_{i,j} F_{ij}\, \tilde u_{l,i}\, \tilde u_{r,j} = 0
\quad\Longleftrightarrow\quad
\underbrace{[\,u_l u_r,\ u_l v_r,\ u_l,\ v_l u_r,\ v_l v_r,\ v_l,\ u_r,\ v_r,\ 1\,]}_{\mathbf{a}^T\ =\ (\tilde{\mathbf{u}}_l \otimes \tilde{\mathbf{u}}_r)^T}
\cdot \underbrace{[F_{11}, F_{12}, \dots, F_{33}]^T}_{\mathbf{f}} = 0 .$$

**Зачем так:** ограничение **линейно по неизвестным** $F_{ij}$ (хотя квадратично по координатам точек).
Каждое соответствие даёт одну строку; $N$ соответствий – систему $A\mathbf{f} = 0$, $A \in \mathbb{R}^{N\times 9}$.

**Почему 8 точек:** у $\mathbf{f}$ 9 элементов, но уравнение однородное – $F$ определена с точностью до масштаба, остаётся 8 неизвестных.
Ровно 8 соответствий дают (в общем положении) единственное с точностью до масштаба решение; больше – переопределённая система, которую
решаем в смысле наименьших квадратов и тем самым усредняем шум.

**Постановка задачи.** У однородной системы всегда есть бесполезное решение $\mathbf{f} = 0$, а любое решение можно умножить на число.
Обе проблемы снимает нормировка $\|\mathbf{f}\| = 1$: она фиксирует масштаб и запрещает ноль.
Кроме того, из-за шума при $N > 8$ точного решения $A\mathbf{f} = 0$ нет, поэтому вместо равенства ищем $\mathbf{f}$, при котором
$A\mathbf{f}$ **ближе всего к нулю**:
$$\mathbf{f}^* = \arg\min_{\mathbf{f}} \|A\mathbf{f}\| \quad \text{при} \quad \|\mathbf{f}\| = 1 .$$

*Это тот же шаблон DLT (Direct Linear Transformation), что при калибровке камеры и оценке гомографии:
«каждое наблюдение – строка, решение – нуль-вектор».* Вспомним, как ставилась задача там:

| | калибровка камеры | гомография | фундаментальная матрица |
|---|---|---|---|
| модель | $\tilde{\mathbf{u}} \simeq P\,\tilde{\mathbf{X}}$, $P \in \mathbb{R}^{3\times 4}$ | $\tilde{\mathbf{u}}' \simeq H\,\tilde{\mathbf{u}}$, $H \in \mathbb{R}^{3\times 3}$ | $\tilde{\mathbf{u}}_l^T F\, \tilde{\mathbf{u}}_r = 0$, $F \in \mathbb{R}^{3\times 3}$ |
| наблюдение | 3D-точка $\leftrightarrow$ пиксель | пиксель $\leftrightarrow$ пиксель | пиксель $\leftrightarrow$ пиксель |
| неизвестные | $\mathbf{p} \in \mathbb{R}^{12}$ (11 с точностью до масштаба) | $\mathbf{h} \in \mathbb{R}^{9}$ (8) | $\mathbf{f} \in \mathbb{R}^{9}$ (8) |
| строк на наблюдение | 2 | 2 | 1 |
| минимум наблюдений | 6 | 4 | 8 |
| задача | $\min \lVert A\mathbf{p}\rVert,\ \lVert\mathbf{p}\rVert=1$ | $\min \lVert A\mathbf{h}\rVert,\ \lVert\mathbf{h}\rVert=1$ | $\min \lVert A\mathbf{f}\rVert,\ \lVert\mathbf{f}\rVert=1$ |

Для $P$ и $H$ «$\simeq$» означает равенство с точностью до неизвестного множителя. Его исключают (например, через
$\tilde{\mathbf{u}} \times P\tilde{\mathbf{X}} = 0$), и из трёх уравнений остаются два независимых. Эпиполярное ограничение уже записано
скалярным равенством без неизвестного множителя, поэтому соответствие даёт всего одну строку, а точек нужно больше.

## 3. Решение через SVD

### 3.1 SVD

**Ортогональные матрицы.** Квадратная матрица $Q$ ортогональна, если $Q^TQ = QQ^T = I$: её столбцы – единичные попарно
перпендикулярные векторы (ортонормированный базис), а обратная матрица – просто транспонированная, $Q^{-1} = Q^T$.
Главное свойство: **ортогональная матрица не меняет длины векторов** (и углы между ними):
$$\|Q\mathbf{x}\|^2 = \mathbf{x}^T Q^T Q\,\mathbf{x} = \mathbf{x}^T\mathbf{x} = \|\mathbf{x}\|^2 .$$
Геометрически это поворот (возможно, с отражением).

**Сингулярное разложение (Singular Value Decomposition).** Любую матрицу $A \in \mathbb{R}^{N\times 9}$ можно записать как
$$\Large A = U\,\Sigma\,V^T,$$
где $U \in \mathbb{R}^{N\times N}$ и $V \in \mathbb{R}^{9\times 9}$ – ортогональные, а $\Sigma \in \mathbb{R}^{N\times 9}$ – диагональная,
с **сингулярными числами** $\sigma_1 \ge \sigma_2 \ge \dots \ge \sigma_9 \ge 0$ на диагонали.
Столбцы $\mathbf{v}_1, \dots, \mathbf{v}_9$ матрицы $V$ – **правые сингулярные векторы**, столбцы $\mathbf{u}_i$ матрицы $U$ – левые:
$A\mathbf{v}_i = \sigma_i \mathbf{u}_i$.

- **Геометрически:** любое линейное отображение – это поворот ($V^T$), растяжение вдоль осей в $\sigma_i$ раз ($\Sigma$) и ещё
  один поворот ($U$). Единичная сфера переходит в эллипсоид с полуосями длины $\sigma_i$; в эти полуоси переходят векторы $\mathbf{v}_i$.
- **Связь с собственными числами:** $A^TA = V\,\Sigma^T\Sigma\,V^T$, т.е. $\mathbf{v}_i$ – собственные векторы $A^TA$,
  а $\sigma_i^2$ – её собственные числа.

### 3.2 Решение

Задача из шага 1: 
$$\Large \min_{\mathbf{f}} \|A\mathbf{f}\|,  \|\mathbf{f}\| = 1.$$

**Её решение – $\mathbf{v}_9$.** 

Перейдём к координатам в базисе $\mathbf{v}_i$: $\mathbf{y} = V^T\mathbf{f}$.
Так как $U$ и $V$ сохраняют длины,
$$\|\mathbf{y}\| = \|\mathbf{f}\| = 1, \qquad \|A\mathbf{f}\| = \|U\,\Sigma\,V^T\mathbf{f}\| = \|\Sigma\,\mathbf{y}\| .$$
Отсюда
$$\|A\mathbf{f}\|^2 = \sum_{i=1}^{9} \sigma_i^2\, y_i^2 \;\ge\; \sigma_9^2 \sum_{i=1}^{9} y_i^2 = \sigma_9^2 .$$
Равенство достигается при $\mathbf{y} = (0, \dots, 0, 1)^T$, т.е. при $\mathbf{f} = V\mathbf{y} = \mathbf{v}_9$.
Значит, решение – **правый сингулярный вектор $A$ с наименьшим сингулярным числом**: последний столбец $V$ = последняя
строка $V^T$ (в numpy: `U, S, Vt = np.linalg.svd(A)`, решение – `Vt[-1]`), а достигнутый минимум $\|A\mathbf{f}^*\| = \sigma_9$.
Без шума $\sigma_9 = 0$ и $\mathbf{v}_9$ – точный нуль-вектор $A$; с шумом $\sigma_9 > 0$.

## 4. Нормировка Хартли

### 4.1. Обусловленность

**Обусловленность** – это чувствительность решения задачи к малым возмущениям входных данных (шуму измерений, ошибкам округления).
Для матрицы её измеряют **числом обусловленности** – отношением наибольшего сингулярного числа к наименьшему:
$$\kappa(A) = \frac{\sigma_{\max}}{\sigma_{\min}} \;\ge\; 1 .$$
Геометрически (шаг 3) это отношение самой длинной полуоси эллипсоида $A\cdot\{\text{единичная сфера}\}$ к самой короткой,
то есть мера того, насколько эллипсоид сплющен.

**Смысл числа:** во сколько раз относительная ошибка входа может усилиться в ответе. Для системы $A\mathbf{x} = \mathbf{b}$
$$\frac{\|\delta\mathbf{x}\|}{\|\mathbf{x}\|} \;\le\; \kappa(A)\,\frac{\|\delta\mathbf{b}\|}{\|\mathbf{b}\|} .$$
- $\kappa \sim 1$–$10$ – **хорошая обусловленность**: ошибка 1% во входе даёт ошибку порядка процентов в ответе.
- $\kappa \gg 1$ – **плохая обусловленность**: задача близка к вырожденной, и крошечная ошибка во входе может полностью
  изменить ответ.

**Почему мерой служат именно сингулярные числа.** Сингулярные числа – это наименьшее и наибольшее растяжение, которое $A$
даёт вектору. Как в шаге 3, положим $\mathbf{y} = V^T\mathbf{x}$; ортогональные $U, V$ не меняют длин, поэтому
$$\|A\mathbf{x}\|^2 = \|\Sigma\,\mathbf{y}\|^2 = \sum_i \sigma_i^2 y_i^2
\quad\Longrightarrow\quad
\sigma_{\min}\|\mathbf{x}\| \;\le\; \|A\mathbf{x}\| \;\le\; \sigma_{\max}\|\mathbf{x}\| .$$
У обратной матрицы $A^{-1} = V\,\Sigma^{-1}U^T$ сингулярные числа равны $1/\sigma_i$: она растягивает не больше чем в $1/\sigma_{\min}$ раз.
Теперь возмутим правую часть: $\delta\mathbf{x} = A^{-1}\delta\mathbf{b}$. Получаем две оценки:
- ошибка усиливается не больше чем в $1/\sigma_{\min}$ раз: $\;\|\delta\mathbf{x}\| \le \|\delta\mathbf{b}\|\,/\,\sigma_{\min}$;
- сам ответ не слишком мал: из $\|\mathbf{b}\| = \|A\mathbf{x}\| \le \sigma_{\max}\|\mathbf{x}\|$ следует
  $\;1/\|\mathbf{x}\| \le \sigma_{\max}/\|\mathbf{b}\|$.

Перемножив их, получаем оценку выше с $\kappa = \sigma_{\max}/\sigma_{\min}$.

- **Оценка достижима.** Худший случай: сигнал лежит в «сильном» направлении ($\mathbf{x} = \mathbf{v}_1$, $\mathbf{b} = \sigma_{\max}\mathbf{u}_1$),
  а шум – в «слабом» ($\delta\mathbf{b} \parallel \mathbf{u}_n$). $A$ сжимает направление $\mathbf{v}_n$ в $\sigma_{\min}$ раз, поэтому
  информация о компоненте $\mathbf{x}$ вдоль него приходит в $\mathbf{b}$ «тихой», шум того же уровня её заглушает, а $A^{-1}$ этот шум
  растягивает в $1/\sigma_{\min}$ раз.
- **Почему отношение, а не просто $\sigma_{\min}$.** Обусловленность описывает *относительные* ошибки и не должна зависеть от масштаба.
  У $A = 0.001\cdot I$ маленькое $\sigma_{\min}$, но это просто смена единиц, и $\kappa = 1$. Плохо не когда растяжения малы, а когда они
  **сильно различаются по направлениям** – эллипсоид сплющен.
- **Почему не собственные числа.** Для несимметричных матриц они не описывают растяжение длин:
  $\begin{bmatrix} 1 & 1000 \\ 0 & 1 \end{bmatrix}$ имеет $\lambda_1 = \lambda_2 = 1$, но $\sigma_1 \approx 1000$, $\sigma_2 \approx 0.001$, т.е. $\kappa \approx 10^6$.
  Для симметричных матриц $\sigma_i = |\lambda_i|$, и разницы нет.
- **Шум в самой матрице** ($A \to A + \delta A$) даёт аналогичную оценку с тем же $\kappa$:
  $\|\delta\mathbf{x}\|/\|\mathbf{x} + \delta\mathbf{x}\| \le \kappa\,\|\delta A\|/\|A\|$. Для нас это главный случай: шум пикселей попадает именно в $A$.

**Пример.** Две почти параллельные прямые $x + y = 2$ и $x + 1.0001\,y = 2.0001$ пересекаются в точке $(1, 1)$.
Сдвинем правую часть второго уравнения на $10^{-4}$ (до $2.0002$) – и точка пересечения уезжает в $(0, 2)$. Здесь $\kappa \approx 4\cdot 10^4$.

**Для однородной задачи** $\min\|A\mathbf{f}\|$, $\|\mathbf{f}\| = 1$ (шаг 3) важно, насколько решение $\mathbf{v}_9$ «отделено» от
остальных направлений. Если $\sigma_9 \ll \sigma_8$, шум $\delta A$ поворачивает решение примерно на угол
$$\angle(\mathbf{f}, \mathbf{f} + \delta\mathbf{f}) \;\approx\; \frac{\|\delta A\|}{\sigma_8}
= \underbrace{\frac{\|\delta A\|}{\sigma_1}}_{\text{относительный шум в } A} \cdot\; \frac{\sigma_1}{\sigma_8} .$$
Роль числа обусловленности здесь играет $\sigma_1/\sigma_8$: во столько раз относительный шум в данных усиливается в ошибку $\mathbf{f}$.

**Последствия плохой обусловленности:**
1. **Усиление шума** – главное. Ошибки измерений в доли пикселя превращаются в большие ошибки ответа. Это свойство самой
   постановки задачи, а не метода: никакой решатель не извлечёт из данных информацию, которой в них нет.
2. **Нестабильность.** Ответ сильно меняется от небольших изменений набора точек: разные случайные выборки дают очень разные решения.
3. **Потеря точности при вычислениях.** В арифметике с плавающей точкой теряется примерно $\log_{10}\kappa$ значащих цифр
   (в `float64` их всего около 16).

**Откуда берётся и как исправить.** Часто плохая обусловленность вызвана не геометрией задачи, а неудачным выбором
**единиц и начала координат**: неизвестные и столбцы матрицы оказываются разного масштаба (как если бы одни величины мерили в метрах,
а другие в микронах). Тогда помогает **нормировка** (предобуславливание) – замена переменных $\mathbf{x} = T\hat{\mathbf{x}}$ с обратимой $T$:
1. перевести данные в «удобные» координаты, где все величины порядка 1 (сдвиг в центр + масштаб);
2. решить хорошо обусловленную задачу в этих координатах;
3. пересчитать ответ обратно в исходные координаты (**денормализация**).

Без шума шаг 3 возвращает то же точное решение. С шумом – решение получается лучше, потому что в нормированных координатах
минимизируется более сбалансированный критерий: все уравнения и все неизвестные вносят сопоставимый вклад.

### 4.2 Нормировка Хартли

**Проблема в нашем случае.** Посмотрите на строку $A$ в шаге 1: элементы отличаются **на 6 порядков**
($u_l u_r \sim 10^6$, $u_l \sim 10^3$, последний – $1$). В таблице ниже видно, что при этом $\sigma_1/\sigma_8 \sim 10^5$. Отсюда три беды:

1. **Плохая обусловленность** (п. 4.1). Шум в 1 пиксель попадает в большие столбцы умноженным на координату:
   $\delta(u_l u_r) \approx u_r\,\delta u_l \sim 10^3$. Этого хватает, чтобы заметно повернуть нуль-вектор $\mathbf{f}$.
2. **Перекос весов.** Сумма квадратов $\|A\mathbf{f}\|^2$ определяется столбцами $\sim 10^6$, а элементы $F$ при маленьких
   столбцах подбираются плохо. Сами элементы настоящей $F$ в пикселях различаются на много порядков, а условие $\|\mathbf{f}\|=1$
   считает их все равноправными.
3. **Нет инвариантности.** Результат зависит от произвольного выбора пиксельной системы координат (где начало – в углу или в центре,
   какой масштаб). Нормировка приводит точки к **канонической** системе – и результат от этого выбора не зависит.

**Нормировка.** Для каждого изображения **отдельно** по его точкам $\mathbf{u}_1, \dots, \mathbf{u}_N$ считаем центр масс и масштаб:
$$\mathbf{c} = \frac1N\sum_i \mathbf{u}_i, \qquad s = \frac{\sqrt 2}{\frac1N\sum_i \|\mathbf{u}_i - \mathbf{c}\|}, \qquad
T = \begin{bmatrix} s & 0 & -s\,c_x\\ 0 & s & -s\,c_y\\ 0 & 0 & 1\end{bmatrix} .$$
$T$ сдвигает центр масс точек в начало координат и масштабирует так, чтобы среднее расстояние до начала было $\sqrt 2$.
Тогда «средняя» точка $\approx (\pm1, \pm1, 1)$, и все однородные координаты одного порядка.

**К чему применяется:** к однородным пиксельным координатам соответствий, левые и правые – со своей матрицей:
$$\hat{\mathbf{u}}_{l,i} = T_l\,\tilde{\mathbf{u}}_{l,i} = \begin{bmatrix} s_l\,(u_{l,i} - c_{l,x}) \\ s_l\,(v_{l,i} - c_{l,y}) \\ 1 \end{bmatrix}, \qquad
\hat{\mathbf{u}}_{r,i} = T_r\,\tilde{\mathbf{u}}_{r,i}, \qquad i = 1, \dots, N .$$
Третья координата остаётся равной 1, поэтому матрица $\hat A$ строится из $\hat{\mathbf{u}}$ ровно так же, как $A$ из $\tilde{\mathbf{u}}$.
$T_l \ne T_r$ – это нормально: $F$ и так связывает две разные системы координат.
$T$ считаются по тем точкам, по которым оцениваем $F$: в RANSAC – заново для каждой выборки из 8 точек.

**Алгоритм с нормировкой:**
$$\{\tilde{\mathbf{u}}_{l,i}, \tilde{\mathbf{u}}_{r,i}\}
\;\xrightarrow{\;T_l,\,T_r\;}\; \{\hat{\mathbf{u}}_{l,i}, \hat{\mathbf{u}}_{r,i}\}
\;\longrightarrow\; \hat A
\;\xrightarrow{\;\text{SVD}\;}\; \hat F
\;\xrightarrow{\;\text{ранг 2}\;}\; \hat F
\;\xrightarrow{\;\text{денормализация}\;}\; F .$$

**Денормализация.** $\hat F$ найдена для нормированных точек: $\hat{\mathbf{u}}_l^T \hat F\, \hat{\mathbf{u}}_r = 0$. Подставим $\hat{\mathbf{u}} = T\tilde{\mathbf{u}}$:
$$\hat{\mathbf{u}}_l^T \hat F\, \hat{\mathbf{u}}_r = (T_l\tilde{\mathbf{u}}_l)^T \hat F\, (T_r\tilde{\mathbf{u}}_r)
= \tilde{\mathbf{u}}_l^T\, \underbrace{T_l^T \hat F\, T_r}_{F}\, \tilde{\mathbf{u}}_r = 0
\qquad\Longrightarrow\qquad F = T_l^T\, \hat F\, T_r .$$
Обратите внимание: здесь **транспонированные $T$, а не обратные** – точки преобразуются матрицей $T$, а $F$ «впитывает» её с двух сторон.
После этого делим $F$ на $\|F\|$: $F$ определена с точностью до масштаба, а $T_l^T \hat F T_r$ уже не единичной нормы.

**Когда денормализовать:**
- **после** зануления $\sigma_3$ (шаг 5): ближайшую матрицу ранга 2 ищем в нормированных координатах, где элементы $\hat F$
  сопоставимы по масштабу. Сам ранг при денормализации сохраняется, так как $T$ обратимы;
- **до** любого использования $F$ с пиксельными точками: эпиполярные линии $F\tilde{\mathbf{u}}_r$, ошибка Сэмпсона в пикселях (шаг 7),
  $E = K^T F K$ (шаг 10).

```python
def normalization_matrix(p):
    c = p.mean(axis=0)
    s = np.sqrt(2) / np.mean(np.linalg.norm(p - c, axis=1))
    return np.array([[s, 0, -s * c[0]],
                     [0, s, -s * c[1]],
                     [0, 0, 1]])


def normalize(u_hat, T):
    return (u_hat @ T.T)[:, :2]


def denormalize(F_hat, T_l, T_r):
    # u_l^T (T_l^T F̂ T_r) u_r = û_l^T F̂ û_r
    F = T_l.T @ F_hat @ T_r
    return F / np.linalg.norm(F)
```

```python
T_l = normalization_matrix(pts_l)
T_r = normalization_matrix(pts_r)
pts_l_n = normalize(to_h(pts_l), T_l)
pts_r_n = normalize(to_h(pts_r), T_r)
A_norm = build_A(pts_l_n, pts_r_n)
```

## 5. Ранг 2: зануление $\sigma_3$

**Что такое $\sigma_3$.** Вектор $\mathbf{f} = \mathbf{v}_9$ из шага 3, записанный матрицей $3\times 3$, – это $\hat F$.
У неё своё SVD, теперь уже для матрицы $3\times 3$:
$$\hat F = U\,\mathrm{diag}(\sigma_1, \sigma_2, \sigma_3)\,V^T, \qquad \sigma_1 \ge \sigma_2 \ge \sigma_3 \ge 0 .$$
$\sigma_3$ – **наименьшее сингулярное число матрицы $\hat F$**.

Так как $U$ и $V$ обратимы, ранг матрицы равен числу ненулевых $\sigma_i$, а $|\det \hat F| = \sigma_1\sigma_2\sigma_3$. Поэтому
$$\operatorname{rank}\hat F = 2 \iff \sigma_3 = 0 \iff \det\hat F = 0 .$$

**Почему $F$ обязана иметь ранг 2:** $F = K^{-T}[\mathbf{t}]_\times R K^{-1}$, а $\operatorname{rank}[\mathbf{t}]_\times = 2$.
Матрица $[\mathbf{t}]_\times$ задаёт векторное произведение, $[\mathbf{t}]_\times\mathbf{x} = \mathbf{t}\times\mathbf{x}$:
$$[\mathbf{t}]_\times = \begin{bmatrix} 0 & -t_3 & t_2\\ t_3 & 0 & -t_1\\ -t_2 & t_1 & 0\end{bmatrix} .$$

- **через определитель:** $[\mathbf{t}]_\times$ кососимметрична, поэтому $\det[\mathbf{t}]_\times = \det[\mathbf{t}]_\times^T = \det(-[\mathbf{t}]_\times) = (-1)^3\det[\mathbf{t}]_\times$,
  откуда $\det = 0$. При этом $[\mathbf{t}]_\times \ne 0$ (есть ненулевые миноры $2\times2$), так что ранг ровно 2.
  Её сингулярные числа – $(\|\mathbf{t}\|, \|\mathbf{t}\|, 0)$.

Это верно при $\mathbf{t} \ne 0$. При чистом вращении камеры $[\mathbf{t}]_\times = 0$, $F$ не определена, а снимки связаны гомографией.
Остальные множители $K^{-T}$, $R$, $K^{-1}$ обратимы, а умножение на обратимые матрицы ранг не меняет, поэтому
$\operatorname{rank}F = \operatorname{rank}[\mathbf{t}]_\times = 2$.

Геометрически это существование **эпиполя**: все эпиполярные линии $\mathbf{l_l} = F\tilde{\mathbf{u}}_r$ проходят через одну точку
$\mathbf{e_l}$, т.е. $\mathbf{e_l}^T F \tilde{\mathbf{u}}_r = 0$ для любого $\tilde{\mathbf{u}}_r$ $\Leftrightarrow$ $F^T\mathbf{e_l} = 0$ $\Rightarrow$ $\det F = 0$.

**Почему линейный алгоритм этого не обеспечивает:** $\det F = 0$ – **кубическое** условие на элементы $\mathbf{f}$, в линейную систему его не вписать.
С шумом решение получается полного ранга ($\sigma_3 > 0$): эпиполя нет, линии не пересекаются в одной точке, а $E = K^TFK$ не имеет вида $[\mathbf{t}]_\times R$.

**Как исправить:** по теореме Эккарта–Янга ближайшая (по норме Фробениуса) матрица ранга 2 получается занулением $\sigma_3$:
$$\hat F = U\,\mathrm{diag}(\sigma_1, \sigma_2, \sigma_3)\,V^T \ \longrightarrow\ \hat F' = U\,\mathrm{diag}(\sigma_1, \sigma_2, 0)\,V^T,
\qquad \|\hat F - \hat F'\|_F = \sigma_3 \ \text{– минимально возможная правка.}$$

** **
1. **Ранг становится ровно 2.** У $\hat F'$ ровно два ненулевых сингулярных числа, поэтому $\det\hat F' = 0$.
2. **Эпиполи появляются и находятся явно.** 
3. **Правка минимальна.**

Делаем это в нормированных координатах, до денормализации: там элементы $\hat F$ сопоставимы по масштабу, и «ближайшая» матрица имеет смысл.

```python
def enforce_rank2(F):
    U, S, Vt = np.linalg.svd(F)
    return U @ np.diag([S[0], S[1], 0]) @ Vt


F_hat = enforce_rank2(F_hat_full)
print("сингулярные числа F̂ до:   ", np.linalg.svd(F_hat_full, compute_uv=False))
print("сингулярные числа F̂ после:", np.linalg.svd(F_hat, compute_uv=False))
```

**Иллюстрация.** Сильнее всего эффект виден при минимальной выборке из 8 точек (так будет внутри RANSAC).
Построим эпиполярные линии на левом изображении для 30 точек и найдём точки пересечения всех пар линий.
При ранге 3 они разбросаны, при ранге 2 – совпадают: это и есть эпиполь.
(Выборка случайная и взята до RANSAC, в ней могут быть выбросы, поэтому сам эпиполь здесь может оказаться где угодно –
даже в кадре, хотя настоящий далеко за ним. Важно лишь, что при ранге 2 он **существует**.)

## 6. Весь алгоритм целиком: восьмиточечный алгоритм оценки F с нормировкой Хартли и занулением $\sigma_3$

```python
def eight_point(p_l, p_r, normalize=True):
    if normalize:
        T_l, T_r = normalization_matrix(p_l), normalization_matrix(p_r)
    else:
        T_l = T_r = np.eye(3)

    A = build_A((to_h(p_l) @ T_l.T)[:, :2], (to_h(p_r) @ T_r.T)[:, :2])   # шаги 1–2
    F_hat = np.linalg.svd(A)[2][-1].reshape(3, 3)                          # шаг 3
    F_hat = enforce_rank2(F_hat)                                           # шаг 4
    return denormalize(F_hat, T_l, T_r)                                    # шаг 5


F_all = eight_point(pts_l, pts_r)
print("F по всем соответствиям (пока с выбросами) =\n", F_all / F_all[2, 2])
```

## Меры ошибки: алгебраическая и Сэмпсона

Чтобы понять, насколько пара точек согласуется с $F$, нужна мера «промаха».

* **Алгебраическая** $\ r = \tilde{\mathbf{u}}_l^T F \tilde{\mathbf{u}}_r$ – её минимизирует сам алгоритм. Без единиц измерения,
  зависит от нормировки $F$ и от системы координат – **порог на неё не поставить**. Место – только внутри решателя.
* **Геометрическая** – расстояние от точки до эпиполярной линии в пикселях: $\ d(\tilde{\mathbf{u}}_r, F^T\tilde{\mathbf{u}}_l) = \dfrac{|\tilde{\mathbf{u}}_l^T F \tilde{\mathbf{u}}_r|}{\sqrt{(F^T\tilde{\mathbf{u}}_l)_1^2 + (F^T\tilde{\mathbf{u}}_l)_2^2}}$.
* **Сэмпсона** – приближение первого порядка к геометрической ошибке, симметричное по двум изображениям:
$$d_{S} = \frac{|\tilde{\mathbf{u}}_l^T F \tilde{\mathbf{u}}_r|}{\sqrt{(F\tilde{\mathbf{u}}_r)_1^2 + (F\tilde{\mathbf{u}}_r)_2^2 + (F^T\tilde{\mathbf{u}}_l)_1^2 + (F^T\tilde{\mathbf{u}}_l)_2^2}} \quad [\text{пикс.}]$$
  В пикселях, в замкнутой форме, дёшево → **порог инлайеров в RANSAC**.
* **Репроекционная** (по 3D-точкам) – самая правильная статистически, но нужна нелинейная оптимизация → для **уточнения** (шаг 11).

```python
def algebraic_err(F, p_l, p_r):
    return np.sum(to_h(p_l) * (to_h(p_r) @ F.T), axis=1)


def sampson_err(F, p_l, p_r):
    ul, ur = to_h(p_l), to_h(p_r)
    Fur = ur @ F.T          # F u_r   (линия на левом изображении)
    Ftul = ul @ F           # F^T u_l (линия на правом изображении)
    num = np.sum(ul * Fur, axis=1) ** 2
    den = Fur[:, 0] ** 2 + Fur[:, 1] ** 2 + Ftul[:, 0] ** 2 + Ftul[:, 1] ** 2
    return np.sqrt(num / den)


d_all = sampson_err(F_all, pts_l, pts_r)
print("алгебраическая ошибка зависит от масштаба F:  "
      f"|r| для F и 1000·F: {np.median(np.abs(algebraic_err(F_all, pts_l, pts_r))):.2e} vs "
      f"{np.median(np.abs(algebraic_err(1000 * F_all, pts_l, pts_r))):.2e}")
print(f"ошибка Сэмпсона от масштаба не зависит: {np.median(d_all):.3f} vs {np.median(sampson_err(1000 * F_all, pts_l, pts_r)):.3f} пикс.")

plt.figure(figsize=(8, 3.2))
plt.hist(np.clip(d_all, 0, 30), bins=90, color="gray")
plt.xlabel("ошибка Сэмпсона, пикс. (обрезано на 30)"); plt.ylabel("число пар")
plt.title(f"F по ВСЕМ соответствиям: медиана {np.median(d_all):.2f} пикс. — выбросы портят МНК-решение")
plt.show()
```

## RANSAC

МНК «тянется» к выбросам: одна грубо неверная пара даёт невязку в сотни пикселей и перекашивает решение.
**RANSAC** (RANdom SAmple Consensus):
1. случайно выбрать **минимальную** выборку – 8 пар;
2. построить по ней гипотезу $F$ восьмиточечным алгоритмом;
3. посчитать, сколько пар согласуется с ней: $d_S < \tau$ (здесь $\tau = 1$ пикс.) – **инлайеры**;
4. повторить $N$ раз, оставить гипотезу с наибольшим числом инлайеров;
5. переоценить $F$ по **всем** её инлайерам (обычный восьмиточечный алгоритм по многим точкам).

**Сколько итераций.** Выборка из $s$ точек целиком состоит из инлайеров с вероятностью $w^s$ ($w$ – доля инлайеров).
Чтобы хотя бы одна из $N$ выборок была «чистой» с вероятностью $p$:
$$N = \frac{\log(1-p)}{\log(1-w^s)} .$$

Число итераций при $p = 0.99$:

| $w$ | $s=5$ ($E$, известна $K$) | $s=7$ ($F$) | $s=8$ ($F$) |
|---|---|---|---|
| 0.9 | 6 | 8 | 9 |
| 0.5 | 146 | 588 | 1177 |
| 0.3 | 1893 | 21 000 | 70 000 |

Отсюда смысл **минимальных решателей**: чем меньше $s$, тем экспоненциально меньше итераций при большой доле выбросов.
Долю $w$ заранее не знаем, поэтому $N$ пересчитываем по ходу – по лучшей найденной гипотезе (адаптивный RANSAC).

```python
def ransac_fundamental(p_l, p_r, thr=1.0, p=0.999, max_iters=5000):
    best_inl, history, n_needed = np.zeros(len(p_l), bool), [], max_iters
    it = 0
    while it < min(n_needed, max_iters):
        idx = rng.choice(len(p_l), 8, replace=False)             # 1. минимальная выборка
        F = eight_point(p_l[idx], p_r[idx])                      # 2. гипотеза
        inl = sampson_err(F, p_l, p_r) < thr                     # 3. инлайеры
        if inl.sum() > best_inl.sum():                           # 4. лучшая гипотеза
            best_inl = inl
            w_est = inl.mean()
            n_needed = np.log(1 - p) / np.log(1 - w_est ** 8)    # адаптивное число итераций
        history.append(best_inl.sum())
        it += 1
    F = eight_point(p_l[best_inl], p_r[best_inl])                # 5. переоценка по всем инлайерам
    inl = sampson_err(F, p_l, p_r) < thr
    return F, inl, np.array(history)


F, inliers, history = ransac_fundamental(pts_l, pts_r)
p_l, p_r = pts_l[inliers], pts_r[inliers]
print(f"итераций: {len(history)},  инлайеров: {inliers.sum()} из {len(inliers)} ({100 * inliers.mean():.1f}%)")
print(f"медиана ошибки Сэмпсона на инлайерах: {np.median(sampson_err(F, p_l, p_r)):.3f} пикс.")
print("F =\n", F / F[2, 2])

plt.figure(figsize=(7, 3))
plt.plot(history); plt.xlabel("итерация"); plt.ylabel("лучшее число инлайеров"); plt.grid(alpha=0.3)
plt.title("адаптивный RANSAC останавливается, когда хорошая гипотеза уже почти наверняка найдена")
plt.show()

show_matches(pts_l, pts_r, colors=np.where(inliers, "lime", "red"), n_max=400,
             title=f"инлайеры (зелёные) и выбросы (красные): {(~inliers).sum()} выбросов")
```

**Зачем нормировка на практике.** Внутри RANSAC гипотезы строятся по 8 точкам – здесь нормировка критична.
Сравним на случайных 8-точечных (и 20-точечных) выборках из инлайеров:

```python
print("медиана ошибки Сэмпсона на всех инлайерах (медиана по 300 случайным выборкам)")
for n in [8, 20]:
    res = {}
    for normalize in [False, True]:
        errs = []
        for _ in range(300):
            idx = rng.choice(len(p_l), n, replace=False)
            errs.append(np.median(sampson_err(eight_point(p_l[idx], p_r[idx], normalize), p_l, p_r)))
        res[normalize] = np.median(errs)
    print(f"  выборка из {n:2d} точек:  без нормировки {res[False]:6.2f} пикс.,  с нормировкой {res[True]:5.2f} пикс.")
```

# Результат: эпиполярные линии

Итоговая $F$ **пиксельные** однородные координаты:
$$\tilde{\mathbf{u}}_l^T F\, \tilde{\mathbf{u}}_r = 0 .$$
Уравнение билинейное: если зафиксировать одну из точек (подставить её числа), оно становится **линейным** по координатам другой,
т.е. задаёт прямую на другом изображении.

- **Линия на левом изображении** – подставляем точку **правого** $\tilde{\mathbf{u}}_r = (u_r, v_r, 1)^T$:
  $$\mathbf{l_l} = F\tilde{\mathbf{u}}_r = \begin{bmatrix} a_l \\ b_l \\ c_l \end{bmatrix}, \qquad
  \tilde{\mathbf{u}}_l^T\mathbf{l_l} = 0 \iff a_l\,u + b_l\,v + c_l = 0 .$$
  На этой прямой (в пикселях левого изображения) должна лежать парная точка $\tilde{\mathbf{u}}_l$.
- **Линия на правом изображении** – подставляем точку **левого** $\tilde{\mathbf{u}}_l = (u_l, v_l, 1)^T$. Так как
  $\tilde{\mathbf{u}}_l^T F\tilde{\mathbf{u}}_r = (F^T\tilde{\mathbf{u}}_l)^T\tilde{\mathbf{u}}_r$,
  $$\mathbf{l_r} = F^T\tilde{\mathbf{u}}_l = \begin{bmatrix} a_r \\ b_r \\ c_r \end{bmatrix}, \qquad a_r\,u + b_r\,v + c_r = 0 .$$


Для рисования из уравнения прямой выражаем $v = -(a\,u + c)/b$ и берём два конца: $u = 0$ и $u = W$ (функция `draw_line`).

**Эпиполи.** Все линии $\mathbf{l_l}$ проходят через одну точку – левый эпиполь $\mathbf{e_l}$: $\mathbf{e_l}^T F\tilde{\mathbf{u}}_r = 0$ для любого
$\tilde{\mathbf{u}}_r$ $\iff$ $F^T\mathbf{e_l} = 0$. Аналогично все $\mathbf{l_r}$ проходят через $\mathbf{e_r}$: $F\mathbf{e_r} = 0$.
Эпиполи – нуль-векторы $F^T$ и $F$, т.е. третьи столбцы $U$ и $V$ в SVD $F$; это проекция центра одной камеры на снимок другой.

```python
def draw_line(ax, l, W, color, lw=1.2):
    a, b, c = l
    ax.plot([0, W], [-c / b, -(a * W + c) / b], color=color, lw=lw)


sel = rng.choice(len(p_l), 12, replace=False)
cols = plt.cm.tab20(np.linspace(0, 1, len(sel)))
fig, ax = plt.subplots(1, 2, figsize=(16, 5.5))
ax[0].imshow(img_l); ax[1].imshow(img_r)
for c, i in zip(cols, sel):
    draw_line(ax[0], F @ np.append(p_r[i], 1), w, c)       # l_l = F u_r
    draw_line(ax[1], F.T @ np.append(p_l[i], 1), w, c)     # l_r = F^T u_l
    ax[0].plot(*p_l[i], "o", color=c, mec="k", ms=7)
    ax[1].plot(*p_r[i], "o", color=c, mec="k", ms=7)
for a, t in zip(ax, ["левое: точки и линии $F u_r$", "правое: точки и линии $F^T u_l$"]):
    a.set_xlim(0, w); a.set_ylim(h, 0); a.set_title(t); a.axis("off")
plt.tight_layout(); plt.show()

e_l = np.linalg.svd(F.T)[2][-1]; e_l /= e_l[2]
e_r = np.linalg.svd(F)[2][-1];   e_r /= e_r[2]
print(f"эпиполи: e_l = ({e_l[0]:.0f}, {e_l[1]:.0f}),  e_r = ({e_r[0]:.0f}, {e_r[1]:.0f})  — далеко за кадром {w}×{h}")
```

# От $F$ к позе: $E$

## SVD -> четыре кандидата

**1. От $F$ к $E$.**
$$E = K^T F K .$$
Это та же эпиполярная связь, только для лучей $\mathbf{a} = K^{-1}\tilde{\mathbf{u}}$ вместо пикселей: $\mathbf{a}_l^T E\,\mathbf{a}_r = 0$.
В $E$ уже нет $K$ – только поза: $E = [\mathbf{t}]_\times R$. Задача – достать из $E$ поворот $R$ и сдвиг $\mathbf{t}$.

**2. Исправляем $E$.** У настоящей $E = [\mathbf{t}]_\times R$ сингулярные числа всегда вида $(\sigma, \sigma, 0)$: поворот $R$ не меняет длин,
поэтому они те же, что у $[\mathbf{t}]_\times$, т.е. $(\|\mathbf{t}\|, \|\mathbf{t}\|, 0)$ (шаг 5). Из-за шума у нашей $E$ это не так,
поэтому заменяем её на ближайшую правильную – так же, как занулением $\sigma_3$ в шаге 5:
$$E = U\,\mathrm{diag}(\sigma_1, \sigma_2, \sigma_3)\,V^T \ \longrightarrow\ U\,\mathrm{diag}(1, 1, 0)\,V^T .$$
Единицы вместо $\sigma$ означают, что берём $\|\mathbf{t}\| = 1$: длину базы по двум снимкам не узнать.

**3. Находим $\mathbf{t}$.** $\mathbf{t}^T E = \mathbf{t}^T[\mathbf{t}]_\times R = 0$, потому что $\mathbf{t}\times\mathbf{t} = 0$. Значит, $\mathbf{t}$ – вектор, который $E$ «обнуляет» слева,
т.е. третий столбец $U$. Знак при этом не определён:
$$\mathbf{t} = \pm\,\mathbf{u}_3 .$$

**4. Находим $R$.** Поворотов, для которых $[\mathbf{t}]_\times R$ даёт нашу $E$, ровно два (проверяется подстановкой):
$$R_1 = U\,W\,V^T, \qquad R_2 = U\,W^T V^T, \qquad
W = \begin{bmatrix} 0 & -1 & 0\\ 1 & 0 & 0\\ 0 & 0 & 1\end{bmatrix}\ \text{(поворот на } 90° \text{ вокруг оси } z\text{)} .$$
$U$ и $V$ должны иметь определитель $+1$, иначе $R$ получится отражением, а не поворотом – поэтому в коде при необходимости меняем их знак.

**5. Откуда четыре кандидата.** 2 варианта $R$ $\times$ 2 знака $\mathbf{t}$ = 4 позы, и все они дают одну и ту же $E$ – то есть одинаково хорошо
объясняют все соответствия. Причина: эпиполярное ограничение проверяет лишь, что два луча и база лежат **в одной плоскости**,
т.е. прямые пересекаются. Где они пересекаются – перед камерами или позади – оно не видит.

Что это за позы:
- $\mathbf{t} \leftrightarrow -\mathbf{t}$ – правая камера стоит по другую сторону от левой;
- $R_1 \leftrightarrow R_2$ – правая камера развёрнута на $180°$ вокруг линии, соединяющей камеры.

Только в одной из четырёх поз точки сцены оказываются **перед обеими камерами**. В остальных они позади одной или обеих
камер – физически так быть не может.

**6. Голосование.** Для каждого кандидата триангулируем **все** инлайеры и считаем точки с положительной глубиной в обеих камерах
($z_l > 0$ и $z_r > 0$). Выбираем кандидата, у которого таких точек **большинство**. В теории хватило бы одной точки, но из-за шума
или малого параллакса отдельная точка может «проголосовать» неверно.

Триангуляция – **метод середины** для всех точек сразу: лучи $\mathbf{a_l} = K^{-1}\tilde{\mathbf{u}}_l$, $\mathbf{d} = R K^{-1}\tilde{\mathbf{u}}_r$,
система $z_l\mathbf{a_l} - z_r\mathbf{d} = \mathbf{t}$ в МНК-смысле (нормальные уравнения 2×2).

```python
E = K.T @ F @ K
U, S, Vt = np.linalg.svd(E)
print("сингулярные числа E / σ1:", S / S[0], " — у существенной матрицы должно быть (1, 1, 0)")
if np.linalg.det(U) < 0:
    U = -U
if np.linalg.det(Vt) < 0:
    Vt = -Vt
E = U @ np.diag([1, 1, 0]) @ Vt                       # проекция на множество существенных матриц
Wm = np.array([[0, -1, 0], [1, 0, 0], [0, 0, 1]])
candidates = [(U @ Wm @ Vt, U[:, 2]), (U @ Wm @ Vt, -U[:, 2]),
              (U @ Wm.T @ Vt, U[:, 2]), (U @ Wm.T @ Vt, -U[:, 2])]


def triangulate_midpoint(p_l, p_r, R, t):
    a = to_h(p_l) @ K_inv.T                       # лучи левой камеры
    d = to_h(p_r) @ K_inv.T @ R.T                 # лучи правой камеры в СК левой: R K^-1 u_r
    # нормальные уравнения для [a, -d] [z_l, z_r]^T = t
    aa, dd, ad = np.sum(a * a, 1), np.sum(d * d, 1), np.sum(a * d, 1)
    at, dt = a @ t, d @ t
    det = aa * dd - ad ** 2
    z_l = (dd * at - ad * dt) / det
    z_r = (ad * at - aa * dt) / det
    P = (z_l[:, None] * a + z_r[:, None] * d + t) / 2   # середина между ближайшими точками лучей
    return P, z_l, z_r


print("\nкандидат   точек перед обеими камерами")
votes = []
for k, (Rc, tc) in enumerate(candidates, 1):
    _, z_l, z_r = triangulate_midpoint(p_l, p_r, Rc, tc)
    votes.append(np.sum((z_l > 0) & (z_r > 0)))
    print(f"    {k}      {votes[-1]:5d} из {len(p_l)}")
R_8pt, t_8pt = candidates[int(np.argmax(votes))]
```

```python
def rot_err_deg(R1, R2):
    return np.degrees(np.arccos(np.clip((np.trace(R1.T @ R2) - 1) / 2, -1, 1)))


def angle_deg(a, b):
    return np.degrees(np.arccos(np.clip(a @ b / np.linalg.norm(a) / np.linalg.norm(b), -1, 1)))


def skew(v):
    # [v]_x: skew(v) @ x = v × x
    return np.array([[0, -v[2], v[1]],
                     [v[2], 0, -v[0]],
                     [-v[1], v[0], 0]])


def F_from_pose(R, t):
    return K_inv.T @ skew(t) @ R @ K_inv


# поза из калибровки: в файле *.camera строки 5–7 – R_w (x_w = R_w x_c + C), строка 8 – центр камеры C в мире
def load_camera(name):
    rows = np.loadtxt(f"assets/fountain-P11/cameras/{name}.png.camera", max_rows=8)
    return rows[4:7], rows[7]


R_wl, C_wl = load_camera("0005")
R_wr, C_wr = load_camera("0003")
# в обозначениях ноутбука x_l = R x_r + t
R_gt = R_wl.T @ R_wr
t_gt = R_wl.T @ (C_wr - C_wl)
print("R_gt =\n", R_gt)
print("t_gt =", t_gt, " м")
print(f"поза после проекции E: ошибка R {rot_err_deg(R_8pt, R_gt):.3f}°,  ошибка направления t {angle_deg(t_8pt, t_gt):.3f}°")
print(f"медиана ошибки Сэмпсона: F из RANSAC {np.median(sampson_err(F, p_l, p_r)):.3f} пикс.,  "
      f"F(R, t) {np.median(sampson_err(F_from_pose(R_8pt, t_8pt), p_l, p_r)):.3f} пикс.")
```

## Уточнение позы

**Зачем.** Сама $F$ хорошая, но **поза, извлечённая из неё, заметно хуже** согласуется с данными: ошибка Сэмпсона для
$F(R,\mathbf{t}) = K^{-T}[\mathbf{t}]_\times RK^{-1}$ выше, чем для $F$ из RANSAC (см. вывод предыдущей ячейки).
Причина: 8-точечный алгоритм подбирает 7 параметров $F$ свободно, часть из них «подстраивается» под шум. Затем проекция
$\mathrm{diag}(\sigma_1,\sigma_2,0) \to \mathrm{diag}(1,1,0)$ насильно делает из неё настоящую $E$ – это ближайшая матрица по Фробениусу,
а не лучшая в пикселях.

Для плотной реконструкции это критично: после ректификации соответствующие точки должны быть на **одной строке** с точностью
до долей пикселя, иначе поиск вдоль строки промахнётся.

**Принцип.** Вместо цепочки «$F$ → $E$ → поза» ищем позу **напрямую**: такие $R$ и $\mathbf{t}$, при которых соответствия лучше всего
согласуются с $F(R,\mathbf{t})$ **в пикселях**:
$$\min_{R,\ \mathbf{t},\ \|\mathbf{t}\|=1}\ \sum_i \rho\Big(d_S\big(\tilde{\mathbf{u}}_{l,i}, \tilde{\mathbf{u}}_{r,i};\ F(R,\mathbf{t})\big)\Big) .$$
Задача нелинейна по $R$ и $\mathbf{t}$, поэтому решаем её итеративно, стартуя с уже найденной позы. Это «bundle adjustment в миниатюре»:
полный bundle adjustment уточняет позы вместе с 3D-точками по ошибке репроекции, а ошибка Сэмпсона позволяет обойтись без 3D-точек.

**Что на входе:**
- `p_l`, `p_r` – пиксельные координаты инлайеров после RANSAC, по $N$ точек;
- $K$ – известная матрица камеры;
- начальное приближение `R_8pt`, `t_8pt` – поза, выбранная голосованием, $\|\mathbf{t}\| = 1$.

**На выходе:** уточнённые `R`, `t` ($\|\mathbf{t}\| = 1$).

**Параметризация: 5 чисел.** Оптимизатор умеет менять только свободные числа, а $R$ должна оставаться поворотом, $\mathbf{t}$ – единичным
вектором. Поэтому позу записываем как **поправку** к начальной $(R_0, \mathbf{t}_0)$ через вектор $\mathbf{x} = (\omega_1, \omega_2, \omega_3, \alpha, \beta)$:
- $R = \mathrm{Rodrigues}(\boldsymbol\omega)\,R_0$ – небольшой дополнительный поворот. $\boldsymbol\omega$ – вектор поворота: направление – ось, длина – угол.
  При любом $\boldsymbol\omega$ получается честный поворот; $\boldsymbol\omega = 0$ даёт $R_0$. **3 параметра.**
- $\mathbf{t} = \mathrm{normalize}(\mathbf{t}_0 + \alpha\mathbf{b}_1 + \beta\mathbf{b}_2)$, где $\mathbf{b}_1, \mathbf{b}_2$ – ортонормированный базис плоскости, перпендикулярной $\mathbf{t}_0$.
  Сдвиг $\mathbf{t}_0$ вбок и нормировка «наклоняют» направление, не меняя длину (длину базы по снимкам всё равно не узнать).
  $\alpha = \beta = 0$ даёт $\mathbf{t}_0$. **2 параметра** – у направления на сфере две степени свободы.

Старт – $\mathbf{x} = 0$, т.е. ровно начальная поза.

**Функция потерь $\rho$ – Хьюбера.** Для невязок меньше `f_scale` = 1 пикс. штраф квадратичный (как обычный МНК), для больших – линейный.
Поэтому оставшиеся после RANSAC выбросы не перетягивают решение на себя.

**Как работает оптимизатор.** `scipy.optimize.least_squares` (по умолчанию метод доверительной области `trf`, родственник Левенберга–Марквардта)
на каждой итерации:
1. считает вектор невязок $\mathbf{r}(\mathbf{x})$ – $N$ ошибок Сэмпсона;
2. численно (конечными разностями) находит якобиан $\partial\mathbf{r}/\partial\mathbf{x}$ размера $N\times 5$;
3. линеаризует задачу вокруг текущего $\mathbf{x}$, решает линейный МНК для шага и ограничивает шаг «доверительной областью»;
4. принимает шаг, если ошибка уменьшилась, и повторяет, пока изменения не станут пренебрежимо малыми.

**Что происходит в коде по шагам:**
1. `pose_residuals(x, R0, t0, b1, b2)` – из 5 чисел `x` собирает $R$ и $\mathbf{t}$ (по формулам выше), строит $F(R,\mathbf{t})$ через `F_from_pose`
   и возвращает вектор ошибок Сэмпсона по всем инлайерам. Именно этот вектор минимизирует оптимизатор.
2. `b1, b2 = np.linalg.svd(t_8pt[None])[2][1:]` – базис плоскости, перпендикулярной $\mathbf{t}_0$. SVD строки $\mathbf{t}_0^T$ (матрица $1\times 3$) даёт
   ортогональную $V^T$: первая строка параллельна $\mathbf{t}_0$, две остальные – ортонормированный базис перпендикулярной плоскости.
3. `least_squares(pose_residuals, np.zeros(5), ..., loss="huber", f_scale=1.0)` – запуск оптимизации из $\mathbf{x} = 0$.
4. Из найденного `res.x` так же, как в `pose_residuals`, собираем итоговые `R` и `t` и нормируем `t`.
5. Таблица и гистограмма: медиана ошибки Сэмпсона и отличие от калибровки для трёх поз – после проекции $E$, после уточнения
   и из калибровки.

**Как читать результат.** После уточнения ошибка Сэмпсона падает на порядок и становится такой же, как у $F$ из RANSAC,
а ошибки $R$ и направления $\mathbf{t}$ относительно калибровки уменьшаются. У позы из калибровки ошибка Сэмпсона может оказаться даже
чуть больше, чем у уточнённой: оптимизация подгоняет позу именно под эти точки, а калибровка тоже измерена с погрешностью.

```python
def pose_residuals(x, R0, t0, b1, b2):
    R = cv2.Rodrigues(x[:3])[0] @ R0
    t = t0 + x[3] * b1 + x[4] * b2
    return sampson_err(F_from_pose(R, t / np.linalg.norm(t)), p_l, p_r)


b1, b2 = np.linalg.svd(t_8pt[None])[2][1:]            # базис плоскости, перпендикулярной t0
res = least_squares(pose_residuals, np.zeros(5), args=(R_8pt, t_8pt, b1, b2), loss="huber", f_scale=1.0)
R = cv2.Rodrigues(res.x[:3])[0] @ R_8pt
t = t_8pt + res.x[3] * b1 + res.x[4] * b2
t /= np.linalg.norm(t)

print("                        Sampson median   ошибка R, °   ошибка t, °")
for name, (Rc, tc) in [("после проекции E", (R_8pt, t_8pt)), ("после уточнения", (R, t)),
                       ("калибровка", (R_gt, t_gt / np.linalg.norm(t_gt)))]:
    print(f"  {name:20s} {np.median(sampson_err(F_from_pose(Rc, tc), p_l, p_r)):10.3f}"
          f"   {rot_err_deg(Rc, R_gt):12.3f}  {angle_deg(tc, t_gt):12.3f}")

bins = np.linspace(0, 5, 60)
plt.figure(figsize=(8, 3.2))
plt.hist(sampson_err(F_from_pose(R_8pt, t_8pt), p_l, p_r), bins=bins, alpha=0.6, label="до уточнения")
plt.hist(sampson_err(F_from_pose(R, t), p_l, p_r), bins=bins, alpha=0.6, label="после уточнения")
plt.xlabel("ошибка Сэмпсона для F(R, t), пикс."); plt.legend(); plt.show()
```

## Триангуляция: разреженное облако

Поза $(R, \mathbf{t})$ известна. Теперь для каждой пары соответствующих точек восстановим 3D-точку сцены: выпустим из обеих камер лучи
через эти пиксели и найдём, где они пересекаются.

**Масштаб: почему $\mathbf{t}$ известен лишь с точностью до длины.** Камера фиксирует только **направления** лучей: пиксель зависит от
отношений $x/z$ и $y/z$, а не от самих координат. Увеличим всю сцену и базу в $\lambda$ раз: $\mathbf{x} \to \lambda\mathbf{x}$, $\mathbf{t} \to \lambda\mathbf{t}$.
Связь $\mathbf{x}_l = R\,\mathbf{x}_r + \mathbf{t}$ сохраняется, а в отношениях $\lambda$ сокращается – **снимки получаются в точности теми же**.
Настоящий фонтан и его макет, снятый камерой, сдвинутой на пропорционально меньшее расстояние, неотличимы.
Поэтому никакой алгоритм не восстановит длину $\mathbf{t}$ по одним изображениям; алгебраически это видно из того, что $E = [\mathbf{t}]_\times R$
определена с точностью до множителя. Мы положили $\|\mathbf{t}\| = 1$.

**Как узнать настоящий масштаб.** Нужна хотя бы одна длина, известная **извне**, в метрах:
- длина базы – стереокамера с известным расстоянием между объективами, калибровка (как здесь: $B$ берём из калибровочных данных);
- известный размер объекта в сцене – шахматная доска, дверной проём, расстояние между двумя отмеченными точками;
- перемещение камеры по GPS, одометрии, IMU;
- высота камеры над землёй (типичный приём в автомобилях), глубина нескольких точек по лидару.

Применяется это одинаково: если реальное расстояние между двумя восстановленными точками $D$ метров, а в реконструкции оно равно $d$,
умножаем все точки и $\mathbf{t}$ на $\lambda = D/d$.

**Что будет при произвольной длине $\mathbf{t}$.** Реконструкция останется правильной **с точностью до подобия**: форма, углы, отношения
расстояний и относительные глубины верные, но все длины умножены на один и тот же неизвестный коэффициент $B_{\text{истинная}}/\|\mathbf{t}\|$.
При $\|\mathbf{t}\| = 1$ глубины получаются «в длинах базы», а не в метрах. От масштаба **не зависят**: ошибка репроекции, углы параллакса,
проверка «точка перед камерами», ректификация – поэтому фильтрация точек ниже сработает так же. Неверными будут только абсолютные величины
(подписи «м» на графиках, пороги в метрах). А если склеивать реконструкции по нескольким парам снимков, у каждой пары будет свой
масштаб, и их придётся согласовывать.

**Как проводится триангуляция (метод середины).** Все координаты – в СК левой камеры.
- **Луч левой камеры** выходит из её центра $O_l = 0$ в направлении $\mathbf{a}_l = K^{-1}\tilde{\mathbf{u}}_l$. Точки луча: $z_l\,\mathbf{a}_l$.
  Третья координата $\mathbf{a}_l$ равна 1, поэтому $z_l$ – это глубина точки в левой камере.
- **Луч правой камеры** выходит из её центра $O_r = \mathbf{t}$ (из $\mathbf{x}_l = R\,\mathbf{x}_r + \mathbf{t}$ при $\mathbf{x}_r = 0$) в направлении
  $\mathbf{d} = R\,K^{-1}\tilde{\mathbf{u}}_r$ (направление, повёрнутое в СК левой). Точки луча: $\mathbf{t} + z_r\,\mathbf{d}$, где $z_r$ – глубина в правой камере.
- **Без шума лучи пересекаются:** $z_l\,\mathbf{a}_l = \mathbf{t} + z_r\,\mathbf{d}$, т.е. $z_l\,\mathbf{a}_l - z_r\,\mathbf{d} = \mathbf{t}$ – 3 уравнения на 2 неизвестные.
  **С шумом** лучи скрещиваются и не пересекаются, поэтому решаем систему в МНК-смысле:
  $$\min_{z_l,\,z_r}\ \big\|z_l\,\mathbf{a}_l - z_r\,\mathbf{d} - \mathbf{t}\big\|^2
  \quad\Longrightarrow\quad
  \begin{bmatrix} \mathbf{a}_l\!\cdot\!\mathbf{a}_l & -\mathbf{a}_l\!\cdot\!\mathbf{d} \\ -\mathbf{a}_l\!\cdot\!\mathbf{d} & \mathbf{d}\!\cdot\!\mathbf{d} \end{bmatrix}
  \begin{bmatrix} z_l \\ z_r \end{bmatrix} =
  \begin{bmatrix} \mathbf{a}_l\!\cdot\!\mathbf{t} \\ -\mathbf{d}\!\cdot\!\mathbf{t} \end{bmatrix}$$
  (нормальные уравнения 2×2, решаем по формулам Крамера сразу для всех точек).
- Найденные $z_l$, $z_r$ дают **ближайшие друг к другу точки** двух лучей: $\mathbf{X}_l = z_l\,\mathbf{a}_l$ и $\mathbf{X}_r = \mathbf{t} + z_r\,\mathbf{d}$.
  3D-точка – **середина** отрезка между ними: $P = (\mathbf{X}_l + \mathbf{X}_r)/2$.
- Знаки $z_l$, $z_r$ заодно говорят, лежит ли точка перед обеими камерами (это и использовалось при голосовании).

Метод середины минимизирует расстояние в 3D, а не ошибку в пикселях. Существуют точнее: линейная DLT-триангуляция, оптимальная
по ошибке репроекции. Но при хорошем параллаксе разница мала.

**Ошибка репроекции.** Берём восстановленную 3D-точку $P$, **проецируем её обратно** в каждую камеру и сравниваем с тем пикселем,
который реально наблюдали.
1. Переводим $P$ в систему координат камеры: для левой $\mathbf{x}_l = P$, для правой $\mathbf{x}_r = R^T(P - \mathbf{t})$
   (из $\mathbf{x}_l = R\,\mathbf{x}_r + \mathbf{t}$). Обозначим координаты точки в камере $(x, y, z)$.
2. Проецируем по модели камеры-обскуры (параметры из $K$):
   $$u' = f_x\,\frac{x}{z} + c_x, \qquad v' = f_y\,\frac{y}{z} + c_y .$$
   Деление на глубину $z$ – перспектива, $f_x, f_y$ переводят в пиксели, $c_x, c_y$ сдвигают начало координат в угол снимка.
3. Ошибка – расстояние в пикселях между спроецированной точкой $(u', v')$ и наблюдённой $(u, v)$:
   $$e = \sqrt{(u' - u)^2 + (v' - v)^2} \quad [\text{пикс.}] .$$

Так считаем $e_l$ для левой камеры и $e_r$ для правой (функция `project`) и для каждой точки берём худшую из двух: $e = \max(e_l, e_r)$.
Смысл: насколько 3D-точка согласуется с **обоими** наблюдениями, в тех же пикселях, что и шум детектора. Лучи, скрещивающиеся из-за шума,
дают середину, которая проецируется чуть мимо наблюдённых точек: доли пикселя – норма. Большая ошибка означает ложное соответствие
или плохую триангуляцию.

**Качество точек.** Оставляем только надёжные:
* **угол параллакса** между лучами в точке $P$ не меньше ~1°: при почти параллельных лучах $\delta Z \approx \frac{Z^2}{fB}\delta d$ – глубина превращается в шум;
* точка перед обеими камерами ($z_l > 0$, $z_r > 0$);
* ошибка репроекции меньше 2 пикселей.

```python
B = np.linalg.norm(t_gt)            # длина базы из калибровки, м (единственное, что берём извне)
t_m = t * B
print(f"база B = {B:.3f} м")

P_sparse, z_l, z_r = triangulate_midpoint(p_l, p_r, R, t_m)

# угол параллакса: между лучами O_l -> P и O_r -> P  (O_r = t в СК левой камеры)
ray_l, ray_r = P_sparse, P_sparse - t_m
parallax = np.degrees(np.arccos(np.clip(np.sum(ray_l * ray_r, 1) /
                                        np.linalg.norm(ray_l, axis=1) / np.linalg.norm(ray_r, axis=1), -1, 1)))


def project(P, R_c, t_c):
    # x_c = R_c^T (x_l - t_c): точка из СК левой камеры в СК камеры с позой (R_c, t_c)
    x = (P - t_c) @ R_c
    x = x @ K.T
    return x[:, :2] / x[:, 2:]


reproj = np.maximum(np.linalg.norm(project(P_sparse, np.eye(3), np.zeros(3)) - p_l, axis=1),
                    np.linalg.norm(project(P_sparse, R, t_m) - p_r, axis=1))
good_pts = (z_l > 0) & (z_r > 0) & (parallax > 1.0) & (reproj < 2.0)
print(f"оставлено {good_pts.sum()} из {len(P_sparse)} точек;  медиана ошибки репроекции {np.median(reproj[good_pts]):.3f} пикс.")

fig, ax = plt.subplots(1, 3, figsize=(18, 4.2))
ax[0].hist(parallax, bins=50, color="gray"); ax[0].axvline(1.0, color="r", ls="--")
ax[0].set_xlabel("угол параллакса, °"); ax[0].set_title("параллакс (порог 1°)")
ax[1].hist(np.clip(reproj, 0, 3), bins=50, color="gray"); ax[1].set_xlabel("ошибка репроекции, пикс.")
ax[1].set_title("ошибка репроекции")
ax[2].imshow(img_l)
sc = ax[2].scatter(p_l[good_pts, 0], p_l[good_pts, 1], c=P_sparse[good_pts, 2], s=5, cmap="turbo")
plt.colorbar(sc, ax=ax[2], label="глубина Z, м"); ax[2].axis("off"); ax[2].set_title("глубина разреженных точек")
plt.tight_layout(); plt.show()
```

```python
Pg = P_sparse[good_pts]
fig, ax = plt.subplots(figsize=(8, 7))
cols_sparse = img_l[p_l[good_pts, 1].astype(int), p_l[good_pts, 0].astype(int)] / 255
ax.scatter(Pg[:, 0], Pg[:, 2], c=cols_sparse, s=4)
for O, Rc, c, n in [(np.zeros(3), np.eye(3), "tab:blue", "$O_l$ (0005)"), (t_m, R, "tab:red", "$O_r$ (0003)")]:
    d = Rc @ np.array([0, 0, 1.0])
    ax.plot(O[0], O[2], "o", color=c, ms=9); ax.arrow(O[0], O[2], d[0], d[2], color=c, width=0.04, head_width=0.2)
    ax.text(O[0] + 0.15, O[2] - 0.5, n, color=c)
zlo, zhi = np.percentile(Pg[:, 2], [1, 99])
ax.set_ylim(-1, zhi + 1)                                  # несколько далёких точек не сжимают график
ax.set_aspect("equal"); ax.grid(alpha=0.3); ax.set_xlabel("x, м"); ax.set_ylabel("z, м")
ax.set_title(f"разреженное облако, вид сверху ({len(Pg)} точек)")
plt.show()
```

# Ректификация

Для **плотной** реконструкции соответствие нужно найти для каждого пикселя. Эпиполярная геометрия сводит поиск к 1D –
вдоль эпиполярной линии. Удобнее всего, когда линии совпадают со **строками**
изображения. **Ректификация** виртуально поворачивает обе камеры вокруг их центров так, чтобы плоскости изображений стали
параллельны базе; изображения пересчитываются гомографиями. После неё $v_l = v_r$, а $d = u_l - u_r$ – **диспаратность**.

`cv2.stereoRectify` ждёт позу второй камеры относительно первой в виде $\mathbf{x}_2 = R_{12}\mathbf{x}_1 + \mathbf{t}_{12}$.
У нас $\mathbf{x_l} = R\mathbf{x_r} + \mathbf{t}$ $\Rightarrow$ $\mathbf{x_r} = R^T\mathbf{x_l} - R^T\mathbf{t}$, т.е. $R_{12} = R^T$, $\mathbf{t}_{12} = -R^T\mathbf{t}$.
`flags=0` разрешает разные главные точки $c_x^L \ne c_x^R$ – для сходящихся камер иначе значительная часть сцены уходит за кадр.

```python
assert t_m[0] > 0, "правая камера должна быть справа от левой (t_x > 0)"
dist0 = np.zeros(5)
RL, RR, PL, PR, Q, _, _ = cv2.stereoRectify(K, dist0, K, dist0, (w, h), R.T, (-R.T @ t_m).reshape(3, 1), flags=0)
mapLx, mapLy = cv2.initUndistortRectifyMap(K, dist0, RL, PL, (w, h), cv2.CV_32FC1)
mapRx, mapRy = cv2.initUndistortRectifyMap(K, dist0, RR, PR, (w, h), cv2.CV_32FC1)
rect_l = cv2.remap(img_l, mapLx, mapLy, cv2.INTER_LINEAR)
rect_r = cv2.remap(img_r, mapRx, mapRy, cv2.INTER_LINEAR)
ones = np.full((h, w), 255, np.uint8)
valid_l = cv2.remap(ones, mapLx, mapLy, cv2.INTER_NEAREST) > 0
valid_r = cv2.remap(ones, mapRx, mapRy, cv2.INTER_NEAREST) > 0

f_rect = PL[0, 0]
B_rect = -PR[0, 3] / f_rect
doffs = PL[0, 2] - PR[0, 2]
print(f"f = {f_rect:.1f} пикс.,  B = {B_rect:.3f} м,  d_offs = c_x^L − c_x^R = {doffs:.1f} пикс.")

q_l = cv2.undistortPoints(p_l.reshape(-1, 1, 2), K, dist0, R=RL, P=PL).reshape(-1, 2)
q_r = cv2.undistortPoints(p_r.reshape(-1, 1, 2), K, dist0, R=RR, P=PR).reshape(-1, 2)

fig, ax = plt.subplots(2, 1, figsize=(16, 11))
for a, (A_, B_), (pa, pb), ttl in [(ax[0], (img_l, img_r), (p_l, p_r), "до ректификации"),
                                   (ax[1], (rect_l, rect_r), (q_l, q_r), "после: соответствующие точки на одной строке")]:
    a.imshow(np.hstack([A_, B_]))
    for y in range(40, h, 70):
        a.axhline(y, color="cyan", lw=0.6, alpha=0.7)
    idx = np.random.default_rng(1).choice(len(pa), 25, replace=False)
    for c, i in zip(plt.cm.hsv(np.linspace(0, 1, len(idx))), idx):
        a.plot([pa[i, 0], pb[i, 0] + w], [pa[i, 1], pb[i, 1]], "-o", color=c, lw=1, ms=4)
    a.set_xlim(0, 2 * w); a.set_ylim(h, 0); a.set_title(ttl); a.axis("off")
plt.tight_layout(); plt.show()

dv = q_l[:, 1] - q_r[:, 1]
print(f"вертикальное расхождение соответствий после ректификации: медиана |Δv| = {np.median(np.abs(dv)):.2f} пикс.")
```

# Плотное сопоставление вдоль строк (SGBM)

**Диспаратность.** После ректификации одна и та же точка сцены видна на обоих снимках **в одной строке** $v$, но в разных столбцах.
Горизонтальный сдвиг между ними и называется **диспаратностью**:
$$d = u_l - u_r \quad [\text{пикс.}] .$$
Это тот же параллакс, что виден, если смотреть на палец перед лицом, закрывая по очереди то один, то другой глаз: близкие предметы
смещаются сильно, далёкие – почти нет. Поэтому диспаратность обратно пропорциональна глубине, $Z = \dfrac{fB}{d - d_{offs}}$ (вывод – в следующем разделе).
У нас главные точки ректифицированных камер различаются (`flags=0`), и сдвиг $d_{offs} = c_x^L - c_x^R$ не равен нулю. Поэтому сами $d$ бывают
и отрицательными; обратно пропорциональна глубине разность $d - d_{offs}$.
**Карта диспаратности** – изображение, в каждом пикселе которого записано $d$. Задача этого раздела – построить её для всех пикселей левого снимка.

**Простейший способ – сопоставление окон по NCC.** Для пикселя $(u, v)$ левого изображения берём окно $W$ вокруг него и сравниваем
с окнами на **той же строке** правого изображения, сдвинутыми на каждое возможное $d$. Мера похожести – **нормированная
взаимная корреляция** (Normalized Cross-Correlation):
$$\mathrm{NCC}(u, v, d) = \frac{\sum_{k \in W}\big(I_L(u+k_u, v+k_v) - \bar I_L\big)\big(I_R(u+k_u-d, v+k_v) - \bar I_R\big)}
{\sqrt{\sum_{k \in W}\big(I_L(u+k_u, v+k_v) - \bar I_L\big)^2\ \sum_{k \in W}\big(I_R(u+k_u-d, v+k_v) - \bar I_R\big)^2}}\ \in [-1, 1],$$
где $\bar I_L$, $\bar I_R$ – средние яркости в окнах. Это коэффициент корреляции между яркостями двух окон: $1$ – окна совпадают с точностью
до яркости и контраста ($I \to aI + b$ не меняет NCC), $0$ – не похожи. Для каждого пикселя **независимо** выбираем $d$ с максимальной NCC
(«победитель забирает всё»).

Проблемы такого **локального** подхода:
- **однородные области** (небо, гладкая стена): все окна похожи, знаменатель близок к нулю, максимум случайный;
- **повторяющаяся текстура** (плитка, окна фасада): несколько одинаково хороших пиков;
- **границы объектов**: окно захватывает и передний план, и фон, поэтому объекты на карте «распухают»;
- **окклюзии**: часть пикселей видна только на одном снимке, у них правильного ответа нет вовсе, но NCC всё равно что-то выберет;
- **размер окна** – компромисс: маленькое шумит, большое размывает детали.

Главное – соседние пиксели решаются независимо, и карта получается зашумлённой, с «выбросами».

**Чем отличается `cv2.StereoSGBM` (Semi-Global Block Matching).**
1. **Стоимость совпадения.** Вместо NCC – сумма по окну `blockSize` метрики Бирчфилда–Томази: это разность яркостей, нечувствительная
   к дискретизации пикселей. Изображение предварительно фильтруется горизонтальной производной (Собель по $x$), что убирает
   влияние общей яркости. Получаем $\mathrm{cost}(p, d)$ – «насколько плохо пикселю $p$ иметь диспаратность $d$».
2. **Гладкость – главное отличие.** Ищем не $d$ для каждого пикселя по отдельности, а всю карту $D$ сразу, минимизируя энергию
   $$E(D) = \sum_p \Big( \mathrm{cost}(p, d_p) + \sum_{q \in N(p)} \big( P_1\,[\,|d_p - d_q| = 1\,] + P_2\,[\,|d_p - d_q| > 1\,] \big) \Big),$$
   где $N(p)$ – соседи пикселя, а $[\cdot]$ равно 1, если условие выполнено, и 0 иначе.
   - Малый штраф $P_1$ – за изменение на 1 пиксель: разрешает наклонные поверхности.
   - Большой штраф $P_2$ – за скачок: разрешает его только там, где его поддерживают данные, т.е. на границах объектов.

   В однородной области стоимости почти одинаковы, и выигрывает гладкое решение – диспаратность «протягивается» от соседей
   с хорошей текстурой. Точная минимизация $E$ по всему 2D-изображению – NP-трудная задача. **Semi-global** означает приближение:
   для каждого пикселя энергия минимизируется динамическим программированием вдоль нескольких 1D-направлений (слева, справа,
   сверху, по диагоналям…), и результаты складываются.
3. **Постобработка** (параметры в коде ниже):
   - `uniquenessRatio` – лучшая стоимость должна быть лучше второй хотя бы на 10%, иначе пиксель помечается как невалидный (неоднозначность);
   - `disp12MaxDiff` – проверка «лево–право»: карта, посчитанная от правого изображения, должна давать то же $d$ (с точностью до 1 пикс.);
     так отсекаются окклюзии;
   - `speckleWindowSize`, `speckleRange` – удаление мелких изолированных «островков» диспаратности (менее 200 пикселей);
   - субпиксельная точность: параболическая интерполяция стоимости вокруг минимума. Поэтому OpenCV хранит $d \cdot 16$ в целых числах.

Итого: NCC решает для каждого пикселя **локально и независимо**, а SGBM учитывает **согласованность соседей** и отбрасывает ненадёжные
пиксели. Карта получается гораздо чище, особенно на однородных участках и границах.
Параметры `P1 = 8·3·BLOCK²`, `P2 = 32·3·BLOCK²` – типичные рекомендованные значения (3 – число цветовых каналов).

**Диапазон поиска** $d$ берём из разреженных соответствий – их диспаратности уже известны.
Техническая деталь: OpenCV не считает диспаратность в левых `MIN_DISP + NUM_DISP` столбцах, поэтому дополняем изображения слева.

```python
disp_sparse = q_l[:, 0] - q_r[:, 0]
lo, hi = np.percentile(disp_sparse, [0.5, 99.5])
margin = 0.1 * (hi - lo) + 8
MIN_DISP = int(np.floor((lo - margin) / 16) * 16)
NUM_DISP = int(np.ceil((hi + margin - MIN_DISP) / 16) * 16)      # кратно 16
PAD = max(0, MIN_DISP + NUM_DISP)
print(f"диспаратности разреженных точек: {lo:.0f} … {hi:.0f}  →  ищем d ∈ [{MIN_DISP}, {MIN_DISP + NUM_DISP})")

BLOCK = 5
sgbm = cv2.StereoSGBM_create(minDisparity=MIN_DISP, numDisparities=NUM_DISP, blockSize=BLOCK,
                             P1=8 * 3 * BLOCK ** 2, P2=32 * 3 * BLOCK ** 2, disp12MaxDiff=1,
                             uniquenessRatio=10, speckleWindowSize=200, speckleRange=2,
                             mode=cv2.STEREO_SGBM_MODE_SGBM_3WAY)
Lp = cv2.copyMakeBorder(rect_l, 0, 0, PAD, 0, cv2.BORDER_CONSTANT, value=0)
Rp = cv2.copyMakeBorder(rect_r, 0, 0, PAD, 0, cv2.BORDER_CONSTANT, value=0)
disp = sgbm.compute(Lp, Rp).astype(np.float32)[:, PAD:] / 16.0     # OpenCV хранит d·16

uu = np.arange(w)[None, :]
ok = (disp >= MIN_DISP) & valid_l & (uu - disp >= 0)
ok &= valid_r[np.arange(h)[:, None], np.clip(np.round(uu - disp).astype(int), 0, w - 1)]
disp[~ok] = np.nan

cmap = plt.get_cmap("turbo").copy(); cmap.set_bad("black")
fig, ax = plt.subplots(1, 2, figsize=(16, 5))
ax[0].imshow(rect_l); ax[0].set_title("левое ректифицированное"); ax[0].axis("off")
im = ax[1].imshow(disp, cmap=cmap, vmin=np.nanpercentile(disp, 1), vmax=np.nanpercentile(disp, 99))
ax[1].set_title(f"диспаратность SGBM (валидно {100 * np.isfinite(disp).mean():.0f}% пикселей)"); ax[1].axis("off")
plt.colorbar(im, ax=ax[1], label="d, пикс."); plt.tight_layout(); plt.show()
```

# Глубина и плотное облако

Для ректифицированной пары ($u_L = c_x^L + fX/Z$, $\ u_R = c_x^R + f(X-B)/Z$):
$$Z = \frac{f\,B}{d - d_{offs}}, \qquad X = \frac{(u - c_x^L)Z}{f}, \qquad Y = \frac{(v - c_y)Z}{f}, \qquad d_{offs} = c_x^L - c_x^R .$$
Это делает `cv2.reprojectImageTo3D` с матрицей $Q$. Точки получаются в СК левой **ректифицированной** камеры;
переводим в СК левой камеры: $\mathbf{x_l} = R_L^T\,\mathbf{x}_{rect}$.

```python
depth = f_rect * B_rect / (disp - doffs)
pts3d = cv2.reprojectImageTo3D(np.nan_to_num(disp, nan=MIN_DISP - 1), Q)
mask = np.isfinite(disp) & (depth > np.nanpercentile(depth, 0.5)) & (depth < np.nanpercentile(depth, 99.5))
X_dense = pts3d[mask].astype(np.float64) @ RL           # = (RL^T x_rect^T)^T  ->  СК левой камеры
C_dense = rect_l[mask]
print(f"плотное облако: {len(X_dense):,} точек (разреженное: {good_pts.sum()})")


zmin, zmax = np.nanpercentile(depth, [1, 99])
fig, ax = plt.subplots(1, 2, figsize=(16, 4.8), gridspec_kw=dict(width_ratios=[1.6, 1]))
im = ax[0].imshow(depth, cmap=cmap.reversed(), vmin=zmin, vmax=zmax)
plt.colorbar(im, ax=ax[0], label="Z, м"); ax[0].set_title("карта глубины"); ax[0].axis("off")

# проверка: глубина SGBM в разреженных точках vs глубина триангуляции (обе – в СК ректифицированной левой камеры)
qi = np.round(q_l[good_pts]).astype(int)
inside = (qi[:, 0] >= 0) & (qi[:, 0] < w) & (qi[:, 1] >= 0) & (qi[:, 1] < h)
z_tri = (P_sparse[good_pts] @ RL.T)[inside, 2]
z_sgbm = depth[qi[inside, 1], qi[inside, 0]]
fin = np.isfinite(z_sgbm)
ax[1].scatter(z_tri[fin], z_sgbm[fin], s=3, alpha=0.5)
ax[1].plot([zmin, zmax], [zmin, zmax], "r--", lw=1)
ax[1].set_xlim(zmin, zmax); ax[1].set_ylim(zmin, zmax); ax[1].set_aspect("equal"); ax[1].grid(alpha=0.3)
ax[1].set_xlabel("Z триангуляции, м"); ax[1].set_ylabel("Z SGBM, м")
ax[1].set_title(f"согласие: медиана |ΔZ| = {np.median(np.abs(z_tri[fin] - z_sgbm[fin])):.3f} м")
plt.tight_layout(); plt.show()
```

```python
def frustum(Rc, tc, color, name, depth_=1.0):
    # пирамида камеры с позой x_l = Rc x_c + tc
    corners = np.array([[0, 0, 1], [w, 0, 1], [w, h, 1], [0, h, 1]], float) @ K_inv.T * depth_
    pts = corners @ Rc.T + tc
    seq = []
    for p in pts:
        seq += [tc, p, [None] * 3]
    seq += [pts[0], pts[1], pts[2], pts[3], pts[0]]
    seq = np.array(seq, dtype=object)
    return go.Scatter3d(x=seq[:, 0], y=seq[:, 1], z=seq[:, 2], mode="lines", name=name, line=dict(color=color, width=4))


idx = rng.choice(len(X_dense), min(80_000, len(X_dense)), replace=False)
Xs, Cs = X_dense[idx].astype(np.float32), C_dense[idx]
fig = go.Figure([
    go.Scatter3d(x=Xs[:, 0], y=Xs[:, 1], z=Xs[:, 2], mode="markers", name="плотное облако",
                 marker=dict(size=1.2, color=[f"rgb({r},{g},{b})" for r, g, b in Cs])),
    frustum(np.eye(3), np.zeros(3), "blue", "левая камера (0005)"),
    frustum(R, t_m, "red", "правая камера (0003)"),
])
fig.update_layout(title=f"плотное облако ({len(idx):,} из {len(X_dense):,} точек)", height=750,
                  margin=dict(l=0, r=0, t=40, b=0), legend=dict(x=0, y=1),
                  scene=dict(aspectmode="data", camera=dict(up=dict(x=0, y=-1, z=0), eye=dict(x=0.3, y=-0.6, z=-1.6))))
fig.show()
```

**Взгляд с новых ракурсов:** проецируем облако в виртуальную камеру с той же $K$, облетающую фонтан (z-buffer: сортировка по глубине,
ближние точки рисуются последними). Белые области – то, чего не видно ни на одном из двух снимков.

```python
def render_cloud(X, colors, splat=2):
    Z = X[:, 2]
    uv = X @ K.T
    u = (uv[:, 0] / Z).astype(int); v = (uv[:, 1] / Z).astype(int)
    ok = (Z > 0.1) & (u >= 0) & (u < w - splat) & (v >= 0) & (v < h - splat)
    order = np.argsort(-Z[ok])
    u, v, c = u[ok][order], v[ok][order], colors[ok][order]
    img = np.full((h, w, 3), 255, np.uint8)
    for dy in range(splat):
        for dx in range(splat):
            img[v + dy, u + dx] = c
    return img


def rot_y(deg):
    a = np.radians(deg)
    return np.array([[np.cos(a), 0, np.sin(a)], [0, 1, 0], [-np.sin(a), 0, np.cos(a)]])


pivot = np.median(X_dense, axis=0)
fig, ax = plt.subplots(2, 3, figsize=(18, 8.5))
for a, ang in zip(ax.flat, [-40, -20, 0, 20, 40]):
    a.imshow(render_cloud((X_dense - pivot) @ rot_y(ang).T + pivot, C_dense))
    a.set_title(f"виртуальная камера: поворот {ang:+d}°"); a.axis("off")
a = ax.flat[-1]
sub = rng.choice(len(X_dense), 200_000, replace=False)
a.scatter(X_dense[sub, 0], X_dense[sub, 2], c=C_dense[sub] / 255, s=0.3)
for O, Rc, c, n in [(np.zeros(3), np.eye(3), "tab:blue", "$O_l$"), (t_m, R, "tab:red", "$O_r$")]:
    d = Rc @ np.array([0, 0, 1.0])
    a.plot(O[0], O[2], "o", color=c, ms=8); a.arrow(O[0], O[2], d[0], d[2], color=c, width=0.05)
    a.text(O[0] + 0.2, O[2] - 0.5, n, color=c)
a.set_aspect("equal"); a.grid(alpha=0.3); a.set_xlabel("x, м"); a.set_ylabel("z, м"); a.set_title("вид сверху")
plt.tight_layout(); plt.show()
```

```python
def save_ply(path, X, colors):
    header = (f"ply\nformat binary_little_endian 1.0\nelement vertex {len(X)}\n"
              "property float x\nproperty float y\nproperty float z\n"
              "property uchar red\nproperty uchar green\nproperty uchar blue\nend_header\n")
    data = np.empty(len(X), dtype=[("x", "<f4"), ("y", "<f4"), ("z", "<f4"), ("r", "u1"), ("g", "u1"), ("b", "u1")])
    data["x"], data["y"], data["z"] = X.T
    data["r"], data["g"], data["b"] = colors.T
    with open(path, "wb") as f:
        f.write(header.encode())
        f.write(data.tobytes())


import os
os.makedirs("outputs", exist_ok=True)
save_ply("outputs/8pt_0005_0003_dense.ply", X_dense, C_dense)
print("сохранено: outputs/8pt_0005_0003_dense.ply — можно открыть в MeshLab / CloudCompare")
```

# Весь пайплайн на OpenCV

Тот же путь «две фотографии → плотное облако», но одним блоком и только функциями `cv2` вместо своих реализаций:

| шаг | своя реализация выше | OpenCV |
|---|---|---|
| соответствия | `cv2.SIFT` + ratio test | то же |
| $F$: 8 точек + нормировка Хартли + ранг 2 + RANSAC | `eight_point`, `ransac_fundamental` | `cv2.findFundamentalMat(..., cv2.FM_RANSAC)` |
| $E$, 4 кандидата, голосование | SVD $E$, `triangulate_midpoint` | `cv2.findEssentialMat` + `cv2.recoverPose` |
| уточнение позы | `least_squares` по ошибке Сэмпсона | локальная оптимизация внутри `cv2.USAC_MAGSAC` |
| триангуляция, репроекция | метод середины, `project` | `cv2.triangulatePoints`, `cv2.projectPoints` |
| ректификация, SGBM, 3D | `cv2.stereoRectify`, `cv2.StereoSGBM` | то же + `cv2.reprojectImageTo3D`, `cv2.transform` |

**Две тонкости.**
- **Конвенция $F$.** OpenCV записывает эпиполярное ограничение как $\tilde{\mathbf{p}}_2^T F\, \tilde{\mathbf{p}}_1 = 0$, где $\mathbf{p}_1$ – точки первого аргумента.
  Мы передаём левые точки первыми, поэтому для нашей записи $\tilde{\mathbf{u}}_l^T F\, \tilde{\mathbf{u}}_r = 0$ матрицу нужно транспонировать.
  Поза из `cv2.recoverPose` тоже «первая → вторая»: $\mathbf{x_r} = R_{12}\mathbf{x_l} + \mathbf{t}_{12}$ – ровно то, что ждёт `cv2.stereoRectify`.
- **Поза – не из $F$.** Если взять $F$ из `cv2.findFundamentalMat`, перейти к $E = K^TFK$ и отдать её в `cv2.recoverPose`, получится та же
  проблема, что в разделе «Уточнение позы»: после ректификации соответствия расходятся по вертикали примерно на 8 пикселей, и SGBM промахивается.
  Поэтому позу берём из `cv2.findEssentialMat`: 5-точечный алгоритм сразу выдаёт настоящую $E$ ($s = 5$ – ещё и меньше итераций RANSAC),
  а MAGSAC уточняет её по всем инлайерам. $F$ из `cv2.findFundamentalMat` используем только для эпиполярных линий.

Масштаб, как и выше, – длина базы из калибровки. Для рисования и сохранения используются функции из предыдущих разделов
(`show_matches`, `draw_line`, `frustum`, `render_cloud`, `rot_y`, `save_ply`). Облако сохраняется в `outputs/8pt_0005_0003_dense_cv2.ply`.

```python
import os
```

```python
cv2.setRNGSeed(0)      # RANSAC в OpenCV случайный – фиксируем зерно

# 0. данные: те же снимки и K (SCALE – из раздела 0)
img_l = cv2.resize(cv2.cvtColor(cv2.imread("assets/fountain-P11/images/0005.png"), cv2.COLOR_BGR2RGB),
                   None, fx=SCALE, fy=SCALE, interpolation=cv2.INTER_AREA)
img_r = cv2.resize(cv2.cvtColor(cv2.imread("assets/fountain-P11/images/0003.png"), cv2.COLOR_BGR2RGB),
                   None, fx=SCALE, fy=SCALE, interpolation=cv2.INTER_AREA)
h, w = img_l.shape[:2]
K = np.diag([SCALE, SCALE, 1.0]) @ np.array([[2759.48, 0.0, 1520.69], [0.0, 2764.16, 1006.81], [0.0, 0.0, 1.0]])
dist0 = np.zeros(5)

# 1. соответствия: SIFT + ratio test
sift = cv2.SIFT_create(nfeatures=8000)
kp_l_cv, des_l_cv = sift.detectAndCompute(cv2.cvtColor(img_l, cv2.COLOR_RGB2GRAY), None)
kp_r_cv, des_r_cv = sift.detectAndCompute(cv2.cvtColor(img_r, cv2.COLOR_RGB2GRAY), None)
knn_cv = cv2.BFMatcher(cv2.NORM_L2).knnMatch(des_l_cv, des_r_cv, k=2)
good_cv = [m for m, n in knn_cv if m.distance < 0.8 * n.distance]
pts_l_cv = np.float64([kp_l_cv[m.queryIdx].pt for m in good_cv])
pts_r_cv = np.float64([kp_r_cv[m.trainIdx].pt for m in good_cv])

# 2. F: нормированный 8-точечный алгоритм внутри RANSAC (порог 1 пикс.)
#    OpenCV пишет p2^T F p1 = 0, поэтому для u_l^T F u_r = 0 транспонируем
F_cv, inl_F_cv = cv2.findFundamentalMat(pts_l_cv, pts_r_cv, cv2.FM_RANSAC, 1.0, 0.999)
F_cv = F_cv.T

# 3. поза: E по 5-точечному алгоритму + MAGSAC, затем выбор 1 из 4 кандидатов голосованием «точки перед камерами»
E_cv, inl_cv = cv2.findEssentialMat(pts_l_cv, pts_r_cv, K, cv2.USAC_MAGSAC, 0.999, 1.0)
_, R12, t12, inl_cv = cv2.recoverPose(E_cv, pts_l_cv, pts_r_cv, K, mask=inl_cv)   # x_r = R12 x_l + t12, |t12| = 1
inl_cv = inl_cv.ravel() > 0
p_l_cv, p_r_cv = pts_l_cv[inl_cv], pts_r_cv[inl_cv]

# масштаб: длина базы из калибровки (8-я строка файла *.camera – центр камеры в мире, м)
C_l, C_r = (np.loadtxt(f"assets/fountain-P11/cameras/{name}.png.camera", skiprows=7, max_rows=1)
            for name in ["0005", "0003"])
t12 = t12 * cv2.norm(C_l - C_r)
R_cv, t_cv = R12.T, (-R12.T @ t12).ravel()               # в обозначениях ноутбука: x_l = R x_r + t

# 4. разреженное облако: DLT-триангуляция, фильтр по глубине и ошибке репроекции
P_l = K @ np.hstack([np.eye(3), np.zeros((3, 1))])
P_r = K @ np.hstack([R12, t12])
X_sparse_cv = cv2.convertPointsFromHomogeneous(cv2.triangulatePoints(P_l, P_r, p_l_cv.T, p_r_cv.T).T).reshape(-1, 3)
rvec_r = cv2.Rodrigues(R12)[0]
reproj_cv = np.maximum(
    np.linalg.norm(cv2.projectPoints(X_sparse_cv, np.zeros(3), np.zeros(3), K, dist0)[0].reshape(-1, 2) - p_l_cv, axis=1),
    np.linalg.norm(cv2.projectPoints(X_sparse_cv, rvec_r, t12, K, dist0)[0].reshape(-1, 2) - p_r_cv, axis=1))
z_r_cv = cv2.transform(X_sparse_cv[:, None], np.hstack([R12, t12]))[:, 0, 2]
good_cv_pts = (X_sparse_cv[:, 2] > 0) & (z_r_cv > 0) & (reproj_cv < 2.0)

# 5. ректификация
RL, RR, PL, PR, Q, _, _ = cv2.stereoRectify(K, dist0, K, dist0, (w, h), R12, t12, flags=0)
mapLx, mapLy = cv2.initUndistortRectifyMap(K, dist0, RL, PL, (w, h), cv2.CV_32FC1)
mapRx, mapRy = cv2.initUndistortRectifyMap(K, dist0, RR, PR, (w, h), cv2.CV_32FC1)
rect_l_cv = cv2.remap(img_l, mapLx, mapLy, cv2.INTER_LINEAR)
rect_r_cv = cv2.remap(img_r, mapRx, mapRy, cv2.INTER_LINEAR)
ones = np.full((h, w), 255, np.uint8)
valid_l = cv2.remap(ones, mapLx, mapLy, cv2.INTER_NEAREST) > 0
valid_r = cv2.remap(ones, mapRx, mapRy, cv2.INTER_NEAREST) > 0
q_l_cv = cv2.undistortPoints(p_l_cv.reshape(-1, 1, 2), K, dist0, R=RL, P=PL).reshape(-1, 2)
q_r_cv = cv2.undistortPoints(p_r_cv.reshape(-1, 1, 2), K, dist0, R=RR, P=PR).reshape(-1, 2)

# 6. плотное сопоставление SGBM; диапазон диспаратностей – по разреженным точкам
lo, hi = np.percentile(q_l_cv[:, 0] - q_r_cv[:, 0], [0.5, 99.5])
margin = 0.1 * (hi - lo) + 8
MIN_DISP = int(np.floor((lo - margin) / 16) * 16)
NUM_DISP = int(np.ceil((hi + margin - MIN_DISP) / 16) * 16)
PAD = max(0, MIN_DISP + NUM_DISP)
BLOCK = 5
sgbm = cv2.StereoSGBM_create(minDisparity=MIN_DISP, numDisparities=NUM_DISP, blockSize=BLOCK,
                             P1=8 * 3 * BLOCK ** 2, P2=32 * 3 * BLOCK ** 2, disp12MaxDiff=1,
                             uniquenessRatio=10, speckleWindowSize=200, speckleRange=2,
                             mode=cv2.STEREO_SGBM_MODE_SGBM_3WAY)
Lp = cv2.copyMakeBorder(rect_l_cv, 0, 0, PAD, 0, cv2.BORDER_CONSTANT, value=0)
Rp = cv2.copyMakeBorder(rect_r_cv, 0, 0, PAD, 0, cv2.BORDER_CONSTANT, value=0)
disp_cv = sgbm.compute(Lp, Rp).astype(np.float32)[:, PAD:] / 16.0
uu = np.arange(w)[None, :]
ok = (disp_cv >= MIN_DISP) & valid_l & (uu - disp_cv >= 0)
ok &= valid_r[np.arange(h)[:, None], np.clip(np.round(uu - disp_cv).astype(int), 0, w - 1)]
disp_cv[~ok] = np.nan

# 7. глубина и плотное облако: reprojectImageTo3D (СК ректифицированной камеры) -> СК левой камеры: x_l = RL^T x_rect
pts3d = cv2.reprojectImageTo3D(np.nan_to_num(disp_cv, nan=MIN_DISP - 1), Q)
depth_cv = np.where(np.isfinite(disp_cv), pts3d[..., 2], np.nan)
mask = np.isfinite(disp_cv) & (depth_cv > np.nanpercentile(depth_cv, 0.5)) & (depth_cv < np.nanpercentile(depth_cv, 99.5))
X_cv = cv2.transform(pts3d[mask][:, None].astype(np.float64), RL.T)[:, 0]
C_cv = rect_l_cv[mask]

os.makedirs("outputs", exist_ok=True)
save_ply("outputs/8pt_0005_0003_dense_cv2.ply", X_cv, C_cv)

print(f"соответствий: {len(pts_l_cv)};  инлайеров F: {int(inl_F_cv.sum())},  инлайеров E: {inl_cv.sum()}")
print(f"R =\n{R_cv}\nt = {t_cv} м")
print(f"разреженное облако: {good_cv_pts.sum()} из {len(X_sparse_cv)} точек,  "
      f"медиана ошибки репроекции {np.median(reproj_cv[good_cv_pts]):.3f} пикс.")
print(f"после ректификации: медиана |Δv| = {np.median(np.abs(q_l_cv[:, 1] - q_r_cv[:, 1])):.2f} пикс.")
print(f"SGBM: d ∈ [{MIN_DISP}, {MIN_DISP + NUM_DISP}),  валидно {100 * np.isfinite(disp_cv).mean():.0f}% пикселей")
print(f"плотное облако: {len(X_cv):,} точек  ->  outputs/8pt_0005_0003_dense_cv2.ply")
```
